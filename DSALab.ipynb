# LAB 1: APPLICATION OF STACK

## Problem:
AI-Based Web Browser History Manager

An AI-powered research assistant used by MSAIML students maintains the recently visited
research pages using a Stack. Since users generally want to return to the most recently
visited page first, the system follows the LIFO (Last In, First Out) principle.

Develop a menu-driven program to implement a Browser History Manager using Stack.
Requirements
Implement the following stack operations:
1. Visit a Page (Push)

  Add a newly visited webpage to the stack.
  Example: Google Scholar → IEEE Xplore → ResearchGate

2. Go Back (Pop)

   Remove and display the most recently visited webpage.
   If the stack is empty, display an appropriate message.

3. Current Page (Peek)

   Display the webpage currently at the top of the stack without removing it.

4. Display History

    Display all webpages currently stored in the stack.

5. Exit

   Terminate the program.

Sample Scenario
1. Visit Google Scholar
2. Visit IEEE Xplore
3. Visit ResearchGate

Current Page → ResearchGate
Go Back
Removed → ResearchGate
Current Page → IEEE Xplore

## Implementation and Demonstration of Operations

```python
# Initialize empty stack and set capacity
history = []
MAX_CAPACITY = 50

def push(page):
    if len(history) >= MAX_CAPACITY:
        print("Stack Overflow: History limit reached.")
    else:
        history.append(page)
        print(f"Visited: {page}")

def pop():
    if len(history) == 0:
        print("Stack Underflow: No history to go back to.")
    else:
        removed = history.pop()
        print(f"Go Back -> Removed: {removed}")

def peek():
    if len(history) == 0:
        print("Current Page: None (History is empty)")
    else:
        print(f"Current Page -> {history[-1]}")

def display():
    if len(history) == 0:
        print("History is empty.")
    else:
        print("\n--- Browsing History ---")
        for i in range(len(history) - 1, -1, -1):
            marker = " (Current)" if i == len(history) - 1 else ""
            print(f"- {history[i]}{marker}")
        print("------------------------")

# --- Step-by-Step Sample Scenario Execution ---
print("=== Running Sample Scenario ===")

# 1. Visit pages
push("Google Scholar")
push("IEEE Xplore")
push("ResearchGate")

# 2. Check current page
peek()

# 3. Go back
pop()

# 4. Check current page again
peek()

print("\n=== Interactive Menu ===")

# Menu-driven loop for further testing
while True:
    print("\n1. Visit a Page (Push)")
    print("2. Go Back (Pop)")
    print("3. Current Page (Peek)")
    print("4. Display History")
    print("5. Exit")

    choice = input("Enter choice (1-5): ")

    if choice == "1":
        page = input("Enter webpage name: ")
        if page != "":
            push(page)
        else:
            print("Page name cannot be empty.")
    elif choice == "2":
        pop()
    elif choice == "3":
        peek()
    elif choice == "4":
        display()
    elif choice == "5":
        print("Exiting program.")
        break
    else:
        print("Invalid choice. Please enter 1 to 5.")
```

## Approach to the Implementation

###Data Representation:

A standard Python dynamic array (list) named history serves as the underlying Stack data structure.

The end of the list (index = -1) represents the top of the stack, while index 0 represents the oldest visited site.

###Stack Boundary Validation:

Overflow Prevention: Defined a fixed boundary (MAX_CAPACITY = 50). Before appending any item, len(history) is evaluated to prevent unbounded memory growth.

Underflow Handling: Before accessing or removing items in pop() and peek(), the length of history is checked against 0 to prevent runtime IndexError exceptions.

###Core Primitive Logic:

Push (push(page)): Appends the webpage to the end of history using built-in .append(), keeping time complexity at $O(1)$.

Pop (pop()): Retrieves and deletes the latest page from the end using .pop(), simulating the browser's "Back" action in $O(1)$ time.

Peek (peek()): Inspects history[-1] directly to show the current active tab without mutating or shrinking the stack.

Display (display()): Traverses history in reverse via range(len(history) - 1, -1, -1) to print pages chronologically from the current active tab down to the first opened page.

###Execution Strategy:

Hardcoded Verification: Runs the exact sequence requested in the lab scenario (Google Scholar $\rightarrow$ IEEE Xplore $\rightarrow$ ResearchGate, followed by peek(), pop(), and a final peek()) to validate functional correctness immediately.

Interactive Testing: Transfers control to an infinite while True menu loop, allowing manual edge-case testing (e.g., popping an empty stack or checking empty history display) via standard input() options.

## Why is LIFO suited for this?

The LIFO (Last In, First Out) property directly models how web browsing navigation works. When you navigate from page A to B to C, the last page visited (C) sits at the top of your attention stack. Hitting the "Back" button should logically return to the immediate predecessor (B), requiring the most recent item to be popped first. A Stack guarantees $O(1)$ constant-time access to push the newest page and pop back to the previous one without shifting or searching through earlier history.

## Complexity Analysis
Time Complexity:

Push: $\mathcal{O}(1)$

Pop: $\mathcal{O}(1)$

Peek: $\mathcal{O}(1)$

Display: $\mathcal{O}(N)$

where $N$ is the number of visited pages currently in the stack.

Space Complexity:

$\mathcal{O}(N)$ auxiliary memory proportional to the number of stored pages.

# LAB 2: IMPLEMENT QUEUE OPERATIONS USING LINKED LIST

## Problem

SmartCare Multispeciality Hospital runs a busy Out-Patient Department (OPD). Every morning hundreds of patients arrive at the registration desk and receive a token. Patients must be called to the doctor strictly in the order in which they registered — first come, first served. The number of patients varies
greatly from day to day (40 on a quiet Sunday, 400+ during a flu season), so the hospital IT team does NOT want a fixed-size array. They have asked you, as a Data Structures intern, to build the core patient-queue module using a singly linked list.

Each patient record (node) must store:

• Token Number — integer, auto-generated in increasing order (101, 102, 103 ...)

• Patient Name — string

• Age — integer

• Department — string (e.g., General Medicine, Orthopaedics, ENT)

• Next — pointer/reference to the next patient in the queue

The queue must maintain two pointers, FRONT (patient to be called next) and REAR (most recently registered patient), so that both insertion and deletion run in O(1) time.

## Approach

Data Structure Selection:

A singly linked list is used so that the queue dynamically grows and shrinks, avoiding the fixed-size limitation of arrays.   

Both FRONT and REAR pointers are maintained to guarantee constant-time $\mathcal{O}(1)$ operations for both Enqueue (insert at REAR) and Dequeue (delete from FRONT).   

Token Generation:

Starts at 101 and increments sequentially with each valid registration.   

Boundary Conditions:

  Queue Underflow:
  
  Triggered when FRONT == NULL (or None) during Dequeue/Peek.   
  First Insertion:
  
  When adding to an empty queue, both FRONT and REAR point to the new node.   
  Single-Node Dequeue:

  When removing the sole remaining element, both FRONT and REAR are reset to NULL (or None).   
  Input Validation:
  
  Zero/negative ages and empty names are rejected.   

```python
class PatientNode:
    def __init__(self, token, name, age, department):
        self.token = token
        self.name = name
        self.age = age
        self.department = department
        self.next = None


class HospitalQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0
        self.next_token = 101

    def is_empty(self):
        return self.front is None

    def enqueue(self, name, age, department):
        if not name or name.strip() == "":
            print("[Error] Patient name cannot be blank.")
            return None
        if age <= 0:
            print("[Error] Age must be a positive integer.")
            return None

        token = self.next_token
        self.next_token += 1

        new_node = PatientNode(token, name.strip(), age, department.strip())

        # First insertion
        if self.is_empty():
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

        self.count += 1
        print(f"Token {token} issued for {name} ({department})")
        return token

    def dequeue(self):
        if self.is_empty():
            print("No patients waiting.")
            return None

        temp = self.front
        print(f"Now calling Token {temp.token} - {temp.name} to {temp.department}")

        self.front = self.front.next
        # Single-node case: reset rear when queue becomes empty
        if self.front is None:
            self.rear = None

        self.count -= 1
        return temp

    def peek(self):
        if self.is_empty():
            print("No patients waiting.")
            return None
        print(f"Token {self.front.token} - {self.front.name} ({self.front.department}, Age: {self.front.age})")
        return self.front

    def display(self):
        if self.is_empty():
            print("No patients waiting.")
            return

        current = self.front
        tokens = []
        while current is not None:
            tokens.append(str(current.token))
            current = current.next

        print(" -> ".join(tokens) + f"; Count = {self.count}")

    def clear(self):
        self.front = None
        self.rear = None
        self.count = 0
        print("All remaining nodes freed.")


def run_part_c_demo():
    print("========================================")
    print("  RUNNING PART C TEST DEMONSTRATION")
    print("========================================")
    q = HospitalQueue()

    print("\n--- Step 1: DEQUEUE on empty queue ---")
    q.dequeue()

    print("\n--- Step 2: ENQUEUE Ravi, 45, General Medicine ---")
    q.enqueue("Ravi", 45, "General Medicine")

    print("\n--- Step 3: ENQUEUE Meena, 32, ENT ---")
    q.enqueue("Meena", 32, "ENT")

    print("\n--- Step 4: ENQUEUE Arjun, 60, Orthopaedics ---")
    q.enqueue("Arjun", 60, "Orthopaedics")

    print("\n--- Step 5: DISPLAY ---")
    q.display()

    print("\n--- Step 6: PEEK ---")
    q.peek()

    print("\n--- Step 7: DEQUEUE ---")
    q.dequeue()

    print("\n--- Step 8: DEQUEUE, DEQUEUE ---")
    q.dequeue()
    q.dequeue()
    print(f"Queue State: FRONT = {q.front}, REAR = {q.rear}")

    print("\n--- Step 9: ENQUEUE Sita, 28, ENT ---")
    q.enqueue("Sita", 28, "ENT")

    print("\n--- Step 10: EXIT ---")
    q.clear()
    print("========================================\n")


def interactive_menu():
    q = HospitalQueue()
    while True:
        print("\n--- SmartCare OPD Queue System ---")
        print("1. Enqueue (Register Patient)")
        print("2. Dequeue (Call Next Patient)")
        print("3. Peek (Next in Line)")
        print("4. Display Waiting List")
        print("5. Check IsEmpty & Count")
        print("6. Run Part C Demonstration Sequence")
        print("7. Exit")
        choice = input("Enter choice (1-7): ").strip()

        if choice == "1":
            name = input("Enter Patient Name: ")
            try:
                age = int(input("Enter Age: "))
            except ValueError:
                print("[Error] Age must be a number.")
                continue
            dept = input("Enter Department: ")
            q.enqueue(name, age, dept)
        elif choice == "2":
            q.dequeue()
        elif choice == "3":
            q.peek()
        elif choice == "4":
            q.display()
        elif choice == "5":
            print(f"Is Empty: {q.is_empty()} | Waiting Patients: {q.count}")
        elif choice == "6":
            run_part_c_demo()
        elif choice == "7":
            q.clear()
            break
        else:
            print("[Error] Invalid menu choice.")


if __name__ == "__main__":
    # Runs the Part C demonstration sequence directly
    run_part_c_demo()


    interactive_menu()
```

# LAB 3: IMPLEMENT STRING MATCHING ALGORITHM

## Problem

GenoScan AI is a bioinformatics start-up building a machine-learning model that predicts gene-regulation activity from DNA sequences. Before the data is fed to the model, a feature-extraction module must locate every occurrence of short, biologically important patterns called motifs (for example, the binding site GCTAG) inside long DNA reads made up only of the characters A, C, G and T.

The number and positions of these motifs become input features for the ML model.
The current prototype uses a brute-force search, which becomes very slow when millions of reads are processed. Repetitive regions such as AAAAAAAA…, which are common in real genomes, make it even worse. You have been hired as a data-structures engineer to replace this module with an efficient, verifiable string-matching component and to justify your choice with evidence.



## Approach

Input Validation:

Both DNA read $T$ and motif $P$ must contain exclusively the uppercase characters A, C, G, and T. If any invalid character is encountered, execution immediately halts with "Invalid input".   

LPS Array Construction ($\mathcal{O}(m)$):

Tracks the length of the Longest Proper Prefix that is also a Suffix for every sub-pattern $P[0 \dots i]$. This dictates where matching resumes after a mismatch without rolling back the text index.   

Naïve Matching ($\mathcal{O}(n \cdot m)$):

Slides a window across all possible starting indices $i \in [0, n - m]$ and evaluates characters linearly until a mismatch occurs.   

KMP Matching ($\mathcal{O}(n + m)$):

Maintains two pointers ($i$ for $T$, $j$ for $P$). When a character matches, both pointers advance. Upon finding a full match or mismatch, $j$ resets to $\text{LPS}[j - 1]$ while $i$ remains untouched, eliminating redundant scans.   

Comparison Instrumentation:

Every character equality test (T[i] == P[j] or T[i+j] == P[j]) increments a dedicated counter. In KMP, the total comparisons are bounded by $2\vert{}T\vert{}$.

```python
def is_valid_dna(s):
    return len(s) > 0 and all(ch in "ACGT" for ch in s)


def compute_lps(pattern):
    m = len(pattern)
    lps = [0] * m
    length = 0
    i = 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps


def naive_search(text, pattern):
    n = len(text)
    m = len(pattern)
    positions = []
    comparisons = 0

    if m > n:
        return positions, comparisons

    for i in range(n - m + 1):
        match = True
        for j in range(m):
            comparisons += 1
            if text[i + j] != pattern[j]:
                match = False
                break
        if match:
            positions.append(i)

    return positions, comparisons


def kmp_search(text, pattern, lps):
    n = len(text)
    m = len(pattern)
    positions = []
    comparisons = 0

    if m > n:
        return positions, comparisons

    i = 0  # index for text
    j = 0  # index for pattern

    while i < n:
        comparisons += 1
        if text[i] == pattern[j]:
            i += 1
            j += 1
            if j == m:
                positions.append(i - m)
                j = lps[j - 1]
        else:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1

    return positions, comparisons


def run_test(text, pattern):
    # Part A Input Validation
    if not is_valid_dna(text) or not is_valid_dna(pattern):
        print("Invalid input")
        return

    lps = compute_lps(pattern)
    naive_pos, naive_comp = naive_search(text, pattern)
    kmp_pos, kmp_comp = kmp_search(text, pattern, lps)

    # Format outputs
    lps_str = " ".join(map(str, lps))
    naive_pos_str = " ".join(map(str, naive_pos)) if naive_pos else "NONE"
    kmp_pos_str = " ".join(map(str, kmp_pos)) if kmp_pos else "NONE"

    print(f"LPS Array: {lps_str}")
    print(f"Naive : Positions = {naive_pos_str} | Count = {len(naive_pos)} | Comparisons = {naive_comp}")
    print(f"KMP   : Positions = {kmp_pos_str} | Count = {len(kmp_pos)} | Comparisons = {kmp_comp}")


def main():
    test_cases = [
        ("ACGTTAGCTAGCTAGGCTAGCTAG", "GCTAG", "TC 1: Normal with overlap"),
        ("AAAAAAAAAA", "AAA", "TC 2: Worst case for Naive"),
        ("ACGTACGT", "TTT", "TC 3: Pattern absent"),
        ("ACG", "ACGTA", "TC 4: |P| > |T|"),
        ("AACGAACGAACG", "AACG", "TC 5: Periodic pattern"),
        ("ATATATATGATAT", "ATAT", "TC 6: Alternating repeats"),
        ("ACGTXACG", "ACG", "TC 7: Invalid input character"),
    ]

    print("=== GenoScan AI - String Matching Demonstration ===")
    for idx, (t, p, desc) in enumerate(test_cases, 1):
        print(f"\n--- Test Case {idx} ({desc}) ---")
        print(f"T = {t}\nP = {p}")
        run_test(t, p)


if __name__ == "__main__":
    main()
```

# LAB 4 : HASHING

## Problem

RailSwift is a start-up that runs ticket-booking counters at a busy railway junction. During festival season, counter agents must retrieve a passenger's booking within milliseconds using the 10-digit PNR (Passenger Name Record) number. The current system stores bookings in an unsorted list and performs a linear search, so lookups slow down sharply as bookings grow, and queues at the counter are getting longer.

You have been hired as a junior engineer to build the core lookup engine as a hash table.

Each booking record contains:

• PNR — 10-digit integer key (e.g. 8203416572); must be unique

• Passenger Name — string

• Train Number — 5-digit integer

• Status — CONFIRMED / RAC / WAITLIST

Because the server has limited memory, the engineering lead wants the table to start small and grow only when necessary. Cancelled tickets must be removed without breaking the lookup of
other bookings.

## Approach

### 1. Open Addressing Design
- **Hash Table Representation**: An array of slots is used, starting with an initial size of $m = 11$.
- **Slot States**: Each slot can be in one of three states:
  - `EMPTY` ($0$)
  - `OCCUPIED` ($1$)
  - `DELETED` ($2$, tombstone)
- **Hash Function**: Division method, defined as:
  $$h(k) = k \pmod m$$
- **Collision Resolution**: Linear probing, defined as:
  $$h(k, i) = (h(k) + i) \pmod m \quad \text{for } i \ge 0$$

### 2. Tombstone Semantics
- **Deletion**: Deleting a record updates the slot status to `DELETED` (tombstone) rather than resetting it to `EMPTY`. If it were reset to `EMPTY`, subsequent lookups for keys that collided at the same home index would terminate prematurely, resulting in false "Not Found" errors.
- **Insertion**: When inserting, if a tombstone is encountered, probing must still continue until an `EMPTY` slot is reached to verify that the key does not already exist in the cluster (preventing duplicate insertion). Once verified as a non-duplicate, the record is placed into the first encountered tombstone slot.

### 3. Dynamic Rehashing
- **Load Factor**: Defined as $\alpha = \frac{n}{m}$, where $n$ is the number of `OCCUPIED` slots.
- **Trigger**: If $\alpha > 0.7$ after an insertion, the table undergoes dynamic resizing.
- **Resizing**: The table size is increased to the smallest prime number $m_{\text{new}} \ge 2m$. Every active `OCCUPIED` record is rehashed using the new modulus, and all `DELETED` tombstones are purged.

### 4. Data Types
- A 10-digit PNR (e.g., $8203416572$) exceeds the standard 32-bit signed integer limit ($2^{31} - 1 \approx 2.14 \times 10^9$). Python handles arbitrarily large integers natively, but in platforms like C/C++, a 64-bit integer (`long long`) must be explicitly used to prevent overflow during calculations.

```python
EMPTY = 0
OCCUPIED = 1
DELETED = 2


def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True


def next_prime(num):
    while not is_prime(num):
        num += 1
    return num


def validate_pnr(pnr):
    return isinstance(pnr, int) and 1000000000 <= pnr <= 9999999999


class RailSwiftHashTable:
    def __init__(self, size=11, enable_rehashing=True):
        self.m = size
        self.enable_rehashing = enable_rehashing
        self.table = [None] * self.m
        self.state = [EMPTY] * self.m
        self.occupied_count = 0

    def hash(self, pnr):
        return pnr % self.m

    def load_factor(self):
        return self.occupied_count / self.m

    def display(self):
        print(f"\n--- Hash Table State (Size = {self.m}, Occupied = {self.occupied_count}, alpha = {self.load_factor():.3f}) ---")
        for i in range(self.m):
            if self.state[i] == OCCUPIED:
                rec = self.table[i]
                print(f"Slot {i:2d}: [OCCUPIED] PNR: {rec['pnr']} | {rec['name']} | Train: {rec['train_no']} | {rec['status']}")
            elif self.state[i] == DELETED:
                print(f"Slot {i:2d}: [DELETED / TOMBSTONE]")
            else:
                print(f"Slot {i:2d}: [EMPTY]")
        print("-" * 75)

    def insert(self, record):
        pnr = record.get("pnr")
        if not validate_pnr(pnr):
            print(f"[Error] Invalid PNR '{pnr}'. PNR must be a positive 10-digit numeric integer.")
            return False

        home_idx = self.hash(pnr)
        first_tombstone_idx = -1
        target_idx = -1
        probes = 0

        # Probe up to m times to avoid infinite loop
        for i in range(self.m):
            idx = (home_idx + i) % self.m
            probes += 1

            if self.state[idx] == OCCUPIED:
                if self.table[idx]["pnr"] == pnr:
                    print(f"[Error] Duplicate PNR: {pnr} already exists at index {idx}.")
                    return False
            elif self.state[idx] == DELETED:
                if first_tombstone_idx == -1:
                    first_tombstone_idx = idx
            elif self.state[idx] == EMPTY:
                target_idx = idx
                break

        # Check for overflow when table is full
        if target_idx == -1 and first_tombstone_idx == -1:
            print("[Error] Hash table overflow: Table is full.")
            return False

        # Reuse tombstone slot if available, else insert into empty slot
        final_idx = first_tombstone_idx if first_tombstone_idx != -1 else target_idx
        self.table[final_idx] = record
        self.state[final_idx] = OCCUPIED
        self.occupied_count += 1

        print(f"Inserted PNR {pnr} -> Home: {home_idx}, Final: {final_idx}, Probes: {probes}")

        # Check load factor and rehash if threshold is exceeded
        if self.enable_rehashing and self.load_factor() > 0.7:
            self.rehash()

        return True

    def search(self, pnr):
        if not validate_pnr(pnr):
            print(f"[Error] Invalid PNR '{pnr}'.")
            return None, 0

        if self.occupied_count == 0:
            print("[Notice] Table is empty.")
            return None, 0

        home_idx = self.hash(pnr)
        probes = 0

        for i in range(self.m):
            idx = (home_idx + i) % self.m
            probes += 1

            if self.state[idx] == EMPTY:
                print(f"Search PNR {pnr} -> Not Found (Stopped at EMPTY slot {idx}, Probes: {probes})")
                return None, probes
            elif self.state[idx] == OCCUPIED and self.table[idx]["pnr"] == pnr:
                rec = self.table[idx]
                print(f"Search PNR {pnr} -> Found at Slot {idx}! ({rec['name']}, {rec['status']}, Probes: {probes})")
                return rec, probes

        print(f"Search PNR {pnr} -> Not Found (Full table probed, Probes: {probes})")
        return None, probes

    def delete(self, pnr):
        if not validate_pnr(pnr):
            print(f"[Error] Invalid PNR '{pnr}'.")
            return False

        if self.occupied_count == 0:
            print("[Notice] Table is empty.")
            return False

        home_idx = self.hash(pnr)
        for i in range(self.m):
            idx = (home_idx + i) % self.m
            if self.state[idx] == EMPTY:
                break
            elif self.state[idx] == OCCUPIED and self.table[idx]["pnr"] == pnr:
                # Mark as DELETED instead of EMPTY to prevent breaking search chains
                self.state[idx] = DELETED
                self.table[idx] = None
                self.occupied_count -= 1
                print(f"Deleted PNR {pnr} at slot {idx}. Slot marked as [DELETED / TOMBSTONE].")
                return True

        print(f"Delete Failed: PNR {pnr} not found.")
        return False

    def rehash(self):
        new_size = next_prime(2 * self.m)
        print(f"\n[!] Load factor {self.load_factor():.3f} exceeded 0.70! Rehashing from size {self.m} to {new_size}...")

        old_table = self.table
        old_state = self.state
        old_m = self.m

        self.m = new_size
        self.table = [None] * self.m
        self.state = [EMPTY] * self.m
        self.occupied_count = 0

        for i in range(old_m):
            if old_state[i] == OCCUPIED:
                # Directly insert into the new table
                self.insert(old_table[i])
        print("[!] Rehashing completed successfully.\n")


# Execution of Mandatory Task 5 Demonstration Steps
def run_task5_demo():
    print("==================================================")
    print("      RAILSWIFT HASH TABLE DEMONSTRATION")
    print("==================================================")
    ht = RailSwiftHashTable(size=11, enable_rehashing=True)

    test_data = [
        {"pnr": 8203416572, "name": "Ananya Rao", "train_no": 12627, "status": "CONFIRMED"},
        {"pnr": 8203416583, "name": "Rahul Menon", "train_no": 12627, "status": "RAC"},
        {"pnr": 8203416540, "name": "Fatima Sheikh", "train_no": 16526, "status": "CONFIRMED"},
        {"pnr": 8203416599, "name": "Vikram Iyer", "train_no": 12008, "status": "WAITLIST"},
        {"pnr": 8203416615, "name": "Divya Nair", "train_no": 12627, "status": "CONFIRMED"},
        {"pnr": 8203416561, "name": "Arjun Patil", "train_no": 16526, "status": "RAC"},
        {"pnr": 8203416604, "name": "Sneha Kulkarni", "train_no": 12008, "status": "CONFIRMED"},
    ]

    print("\n--- Inserting Initial 7 Bookings ---")
    for rec in test_data:
        ht.insert(rec)

    print("\n--- Step 1: Display Table and Load Factor after Insertion #7 ---")
    ht.display()
    print(f"Current Load Factor: {ht.load_factor():.3f}")

    print("\n--- Step 2: Delete PNR 8203416583 (Rahul Menon) ---")
    ht.delete(8203416583)
    ht.display()

    print("\n--- Step 3: Search PNR 8203416561 (Arjun Patil) ---")
    ht.search(8203416561)

    print("\n--- Step 4: Search Deleted PNR 8203416583 ---")
    ht.search(8203416583)

    print("\n--- Step 5: Insert Record #8 (PNR 8203416626: Karthik Gowda) ---")
    record_8 = {"pnr": 8203416626, "name": "Karthik Gowda", "train_no": 12627, "status": "WAITLIST"}
    ht.insert(record_8)
    ht.display()

    print("\n--- Step 6: Demonstrating Boundary Conditions B1 - B7 ---")
    # B1: Duplicate PNR
    print("\n[B1 Check: Duplicate PNR]")
    ht.insert({"pnr": 8203416572, "name": "Duplicate User", "train_no": 12627, "status": "CONFIRMED"})

    # B2: Search/Delete on Empty Table
    print("\n[B2 Check: Empty Table]")
    empty_ht = RailSwiftHashTable(size=11)
    empty_ht.search(8203416572)
    empty_ht.delete(8203416572)

    # B3: Delete Non-Existent PNR
    print("\n[B3 Check: Delete Non-Existent PNR]")
    ht.delete(9999999999)

    # B4: Invalid PNR (Negative, non-10 digit)
    print("\n[B4 Check: Invalid PNR format]")
    ht.insert({"pnr": -12345, "name": "Invalid PNR", "train_no": 12345, "status": "RAC"})
    ht.insert({"pnr": 123, "name": "Short PNR", "train_no": 12345, "status": "RAC"})

    # B5: Search key beyond DELETED slot
    print("\n[B5 Check: Key beyond DELETED slot]")
    print("Verified in Step 3 where PNR 8203416561 was found across slot 0 (tombstone).")

    # B6: Insert reusing tombstone slot
    print("\n[B6 Check: Tombstone slot reuse]")
    print("Verified in Step 5 where PNR 8203416626 reused slot 0 after validating no duplicates.")

    # B7: Hash Table Overflow with Rehashing Disabled
    print("\n[B7 Check: Hash table overflow with rehashing off]")
    small_ht = RailSwiftHashTable(size=3, enable_rehashing=False)
    small_ht.insert({"pnr": 1000000001, "name": "User 1", "train_no": 11111, "status": "CONFIRMED"})
    small_ht.insert({"pnr": 1000000002, "name": "User 2", "train_no": 11111, "status": "CONFIRMED"})
    small_ht.insert({"pnr": 1000000003, "name": "User 3", "train_no": 11111, "status": "CONFIRMED"})
    small_ht.insert({"pnr": 1000000004, "name": "User 4", "train_no": 11111, "status": "CONFIRMED"})


if __name__ == "__main__":
    run_task5_demo()
```
