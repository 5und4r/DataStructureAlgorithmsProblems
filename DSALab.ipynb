# LAB 1: APPLICATION OF STACK

## Problem:
AI-Based Web Browser History Manager

An AI-powered research assistant used by MSAIML students maintains the recently visited
research pages using a Stack. Since users generally want to return to the most recently
visited page first, the system follows the LIFO (Last In, First Out) principle.

Develop a menu-driven program to implement a Browser History Manager using Stack.
Requirements
Implement the following stack operations:
1. Visit a Page (Push)

  Add a newly visited webpage to the stack.
  Example: Google Scholar → IEEE Xplore → ResearchGate

2. Go Back (Pop)

   Remove and display the most recently visited webpage.
   If the stack is empty, display an appropriate message.

3. Current Page (Peek)

   Display the webpage currently at the top of the stack without removing it.

4. Display History

    Display all webpages currently stored in the stack.

5. Exit

   Terminate the program.

Sample Scenario
1. Visit Google Scholar
2. Visit IEEE Xplore
3. Visit ResearchGate

Current Page → ResearchGate
Go Back
Removed → ResearchGate
Current Page → IEEE Xplore

## Implementation and Demonstration of Operations

```python
# Initialize empty stack and set capacity
history = []
MAX_CAPACITY = 50

def push(page):
    if len(history) >= MAX_CAPACITY:
        print("Stack Overflow: History limit reached.")
    else:
        history.append(page)
        print(f"Visited: {page}")

def pop():
    if len(history) == 0:
        print("Stack Underflow: No history to go back to.")
    else:
        removed = history.pop()
        print(f"Go Back -> Removed: {removed}")

def peek():
    if len(history) == 0:
        print("Current Page: None (History is empty)")
    else:
        print(f"Current Page -> {history[-1]}")

def display():
    if len(history) == 0:
        print("History is empty.")
    else:
        print("\n--- Browsing History ---")
        for i in range(len(history) - 1, -1, -1):
            marker = " (Current)" if i == len(history) - 1 else ""
            print(f"- {history[i]}{marker}")
        print("------------------------")

# --- Step-by-Step Sample Scenario Execution ---
print("=== Running Sample Scenario ===")

# 1. Visit pages
push("Google Scholar")
push("IEEE Xplore")
push("ResearchGate")

# 2. Check current page
peek()

# 3. Go back
pop()

# 4. Check current page again
peek()

print("\n=== Interactive Menu ===")

# Menu-driven loop for further testing
while True:
    print("\n1. Visit a Page (Push)")
    print("2. Go Back (Pop)")
    print("3. Current Page (Peek)")
    print("4. Display History")
    print("5. Exit")

    choice = input("Enter choice (1-5): ")

    if choice == "1":
        page = input("Enter webpage name: ")
        if page != "":
            push(page)
        else:
            print("Page name cannot be empty.")
    elif choice == "2":
        pop()
    elif choice == "3":
        peek()
    elif choice == "4":
        display()
    elif choice == "5":
        print("Exiting program.")
        break
    else:
        print("Invalid choice. Please enter 1 to 5.")
```

## Approach to the Implementation

###Data Representation:

A standard Python dynamic array (list) named history serves as the underlying Stack data structure.

The end of the list (index = -1) represents the top of the stack, while index 0 represents the oldest visited site.

###Stack Boundary Validation:

Overflow Prevention: Defined a fixed boundary (MAX_CAPACITY = 50). Before appending any item, len(history) is evaluated to prevent unbounded memory growth.

Underflow Handling: Before accessing or removing items in pop() and peek(), the length of history is checked against 0 to prevent runtime IndexError exceptions.

###Core Primitive Logic:

Push (push(page)): Appends the webpage to the end of history using built-in .append(), keeping time complexity at $O(1)$.

Pop (pop()): Retrieves and deletes the latest page from the end using .pop(), simulating the browser's "Back" action in $O(1)$ time.

Peek (peek()): Inspects history[-1] directly to show the current active tab without mutating or shrinking the stack.

Display (display()): Traverses history in reverse via range(len(history) - 1, -1, -1) to print pages chronologically from the current active tab down to the first opened page.

###Execution Strategy:

Hardcoded Verification: Runs the exact sequence requested in the lab scenario (Google Scholar $\rightarrow$ IEEE Xplore $\rightarrow$ ResearchGate, followed by peek(), pop(), and a final peek()) to validate functional correctness immediately.

Interactive Testing: Transfers control to an infinite while True menu loop, allowing manual edge-case testing (e.g., popping an empty stack or checking empty history display) via standard input() options.

## Why is LIFO suited for this?

The LIFO (Last In, First Out) property directly models how web browsing navigation works. When you navigate from page A to B to C, the last page visited (C) sits at the top of your attention stack. Hitting the "Back" button should logically return to the immediate predecessor (B), requiring the most recent item to be popped first. A Stack guarantees $O(1)$ constant-time access to push the newest page and pop back to the previous one without shifting or searching through earlier history.

## Complexity Analysis
Time Complexity:

Push: $\mathcal{O}(1)$

Pop: $\mathcal{O}(1)$

Peek: $\mathcal{O}(1)$

Display: $\mathcal{O}(N)$

where $N$ is the number of visited pages currently in the stack.

Space Complexity:

$\mathcal{O}(N)$ auxiliary memory proportional to the number of stored pages.

# LAB 2: IMPLEMENT QUEUE OPERATIONS USING LINKED LIST

## Problem

SmartCare Multispeciality Hospital runs a busy Out-Patient Department (OPD). Every morning hundreds of patients arrive at the registration desk and receive a token. Patients must be called to the doctor strictly in the order in which they registered — first come, first served. The number of patients varies
greatly from day to day (40 on a quiet Sunday, 400+ during a flu season), so the hospital IT team does NOT want a fixed-size array. They have asked you, as a Data Structures intern, to build the core patient-queue module using a singly linked list.

Each patient record (node) must store:

• Token Number — integer, auto-generated in increasing order (101, 102, 103 ...)

• Patient Name — string

• Age — integer

• Department — string (e.g., General Medicine, Orthopaedics, ENT)

• Next — pointer/reference to the next patient in the queue

The queue must maintain two pointers, FRONT (patient to be called next) and REAR (most recently registered patient), so that both insertion and deletion run in O(1) time.

## Approach

Data Structure Selection:

A singly linked list is used so that the queue dynamically grows and shrinks, avoiding the fixed-size limitation of arrays.   

Both FRONT and REAR pointers are maintained to guarantee constant-time $\mathcal{O}(1)$ operations for both Enqueue (insert at REAR) and Dequeue (delete from FRONT).   

Token Generation:

Starts at 101 and increments sequentially with each valid registration.   

Boundary Conditions:

  Queue Underflow:
  
  Triggered when FRONT == NULL (or None) during Dequeue/Peek.   
  First Insertion:
  
  When adding to an empty queue, both FRONT and REAR point to the new node.   
  Single-Node Dequeue:

  When removing the sole remaining element, both FRONT and REAR are reset to NULL (or None).   
  Input Validation:
  
  Zero/negative ages and empty names are rejected.   

```python
class PatientNode:
    def __init__(self, token, name, age, department):
        self.token = token
        self.name = name
        self.age = age
        self.department = department
        self.next = None


class HospitalQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0
        self.next_token = 101

    def is_empty(self):
        return self.front is None

    def enqueue(self, name, age, department):
        if not name or name.strip() == "":
            print("[Error] Patient name cannot be blank.")
            return None
        if age <= 0:
            print("[Error] Age must be a positive integer.")
            return None

        token = self.next_token
        self.next_token += 1

        new_node = PatientNode(token, name.strip(), age, department.strip())

        # First insertion
        if self.is_empty():
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

        self.count += 1
        print(f"Token {token} issued for {name} ({department})")
        return token

    def dequeue(self):
        if self.is_empty():
            print("No patients waiting.")
            return None

        temp = self.front
        print(f"Now calling Token {temp.token} - {temp.name} to {temp.department}")

        self.front = self.front.next
        # Single-node case: reset rear when queue becomes empty
        if self.front is None:
            self.rear = None

        self.count -= 1
        return temp

    def peek(self):
        if self.is_empty():
            print("No patients waiting.")
            return None
        print(f"Token {self.front.token} - {self.front.name} ({self.front.department}, Age: {self.front.age})")
        return self.front

    def display(self):
        if self.is_empty():
            print("No patients waiting.")
            return

        current = self.front
        tokens = []
        while current is not None:
            tokens.append(str(current.token))
            current = current.next

        print(" -> ".join(tokens) + f"; Count = {self.count}")

    def clear(self):
        self.front = None
        self.rear = None
        self.count = 0
        print("All remaining nodes freed.")


def run_part_c_demo():
    print("========================================")
    print("  RUNNING PART C TEST DEMONSTRATION")
    print("========================================")
    q = HospitalQueue()

    print("\n--- Step 1: DEQUEUE on empty queue ---")
    q.dequeue()

    print("\n--- Step 2: ENQUEUE Ravi, 45, General Medicine ---")
    q.enqueue("Ravi", 45, "General Medicine")

    print("\n--- Step 3: ENQUEUE Meena, 32, ENT ---")
    q.enqueue("Meena", 32, "ENT")

    print("\n--- Step 4: ENQUEUE Arjun, 60, Orthopaedics ---")
    q.enqueue("Arjun", 60, "Orthopaedics")

    print("\n--- Step 5: DISPLAY ---")
    q.display()

    print("\n--- Step 6: PEEK ---")
    q.peek()

    print("\n--- Step 7: DEQUEUE ---")
    q.dequeue()

    print("\n--- Step 8: DEQUEUE, DEQUEUE ---")
    q.dequeue()
    q.dequeue()
    print(f"Queue State: FRONT = {q.front}, REAR = {q.rear}")

    print("\n--- Step 9: ENQUEUE Sita, 28, ENT ---")
    q.enqueue("Sita", 28, "ENT")

    print("\n--- Step 10: EXIT ---")
    q.clear()
    print("========================================\n")


def interactive_menu():
    q = HospitalQueue()
    while True:
        print("\n--- SmartCare OPD Queue System ---")
        print("1. Enqueue (Register Patient)")
        print("2. Dequeue (Call Next Patient)")
        print("3. Peek (Next in Line)")
        print("4. Display Waiting List")
        print("5. Check IsEmpty & Count")
        print("6. Run Part C Demonstration Sequence")
        print("7. Exit")
        choice = input("Enter choice (1-7): ").strip()

        if choice == "1":
            name = input("Enter Patient Name: ")
            try:
                age = int(input("Enter Age: "))
            except ValueError:
                print("[Error] Age must be a number.")
                continue
            dept = input("Enter Department: ")
            q.enqueue(name, age, dept)
        elif choice == "2":
            q.dequeue()
        elif choice == "3":
            q.peek()
        elif choice == "4":
            q.display()
        elif choice == "5":
            print(f"Is Empty: {q.is_empty()} | Waiting Patients: {q.count}")
        elif choice == "6":
            run_part_c_demo()
        elif choice == "7":
            q.clear()
            break
        else:
            print("[Error] Invalid menu choice.")


if __name__ == "__main__":
    # Runs the Part C demonstration sequence directly
    run_part_c_demo()


    interactive_menu()
```

# LAB 3: IMPLEMENT STRING MATCHING ALGORITHM

## Problem

GenoScan AI is a bioinformatics start-up building a machine-learning model that predicts gene-regulation activity from DNA sequences. Before the data is fed to the model, a feature-extraction module must locate every occurrence of short, biologically important patterns called motifs (for example, the binding site GCTAG) inside long DNA reads made up only of the characters A, C, G and T.

The number and positions of these motifs become input features for the ML model.
The current prototype uses a brute-force search, which becomes very slow when millions of reads are processed. Repetitive regions such as AAAAAAAA…, which are common in real genomes, make it even worse. You have been hired as a data-structures engineer to replace this module with an efficient, verifiable string-matching component and to justify your choice with evidence.



## Approach

Input Validation:

Both DNA read $T$ and motif $P$ must contain exclusively the uppercase characters A, C, G, and T. If any invalid character is encountered, execution immediately halts with "Invalid input".   

LPS Array Construction ($\mathcal{O}(m)$):

Tracks the length of the Longest Proper Prefix that is also a Suffix for every sub-pattern $P[0 \dots i]$. This dictates where matching resumes after a mismatch without rolling back the text index.   

Naïve Matching ($\mathcal{O}(n \cdot m)$):

Slides a window across all possible starting indices $i \in [0, n - m]$ and evaluates characters linearly until a mismatch occurs.   

KMP Matching ($\mathcal{O}(n + m)$):

Maintains two pointers ($i$ for $T$, $j$ for $P$). When a character matches, both pointers advance. Upon finding a full match or mismatch, $j$ resets to $\text{LPS}[j - 1]$ while $i$ remains untouched, eliminating redundant scans.   

Comparison Instrumentation:

Every character equality test (T[i] == P[j] or T[i+j] == P[j]) increments a dedicated counter. In KMP, the total comparisons are bounded by $2\vert{}T\vert{}$.

```python
def is_valid_dna(s):
    return len(s) > 0 and all(ch in "ACGT" for ch in s)


def compute_lps(pattern):
    m = len(pattern)
    lps = [0] * m
    length = 0
    i = 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps


def naive_search(text, pattern):
    n = len(text)
    m = len(pattern)
    positions = []
    comparisons = 0

    if m > n:
        return positions, comparisons

    for i in range(n - m + 1):
        match = True
        for j in range(m):
            comparisons += 1
            if text[i + j] != pattern[j]:
                match = False
                break
        if match:
            positions.append(i)

    return positions, comparisons


def kmp_search(text, pattern, lps):
    n = len(text)
    m = len(pattern)
    positions = []
    comparisons = 0

    if m > n:
        return positions, comparisons

    i = 0  # index for text
    j = 0  # index for pattern

    while i < n:
        comparisons += 1
        if text[i] == pattern[j]:
            i += 1
            j += 1
            if j == m:
                positions.append(i - m)
                j = lps[j - 1]
        else:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1

    return positions, comparisons


def run_test(text, pattern):
    # Part A Input Validation
    if not is_valid_dna(text) or not is_valid_dna(pattern):
        print("Invalid input")
        return

    lps = compute_lps(pattern)
    naive_pos, naive_comp = naive_search(text, pattern)
    kmp_pos, kmp_comp = kmp_search(text, pattern, lps)

    # Format outputs
    lps_str = " ".join(map(str, lps))
    naive_pos_str = " ".join(map(str, naive_pos)) if naive_pos else "NONE"
    kmp_pos_str = " ".join(map(str, kmp_pos)) if kmp_pos else "NONE"

    print(f"LPS Array: {lps_str}")
    print(f"Naive : Positions = {naive_pos_str} | Count = {len(naive_pos)} | Comparisons = {naive_comp}")
    print(f"KMP   : Positions = {kmp_pos_str} | Count = {len(kmp_pos)} | Comparisons = {kmp_comp}")


def main():
    test_cases = [
        ("ACGTTAGCTAGCTAGGCTAGCTAG", "GCTAG", "TC 1: Normal with overlap"),
        ("AAAAAAAAAA", "AAA", "TC 2: Worst case for Naive"),
        ("ACGTACGT", "TTT", "TC 3: Pattern absent"),
        ("ACG", "ACGTA", "TC 4: |P| > |T|"),
        ("AACGAACGAACG", "AACG", "TC 5: Periodic pattern"),
        ("ATATATATGATAT", "ATAT", "TC 6: Alternating repeats"),
        ("ACGTXACG", "ACG", "TC 7: Invalid input character"),
    ]

    print("=== GenoScan AI - String Matching Demonstration ===")
    for idx, (t, p, desc) in enumerate(test_cases, 1):
        print(f"\n--- Test Case {idx} ({desc}) ---")
        print(f"T = {t}\nP = {p}")
        run_test(t, p)


if __name__ == "__main__":
    main()
```

```python
# Interactive testing cell for DNA pattern matching
def interactive_dna_test():
    print("--- GenoScan AI Custom Search ---")
    text = input("Enter DNA Sequence (Text containing only A, C, G, T): ").strip().upper()
    pattern = input("Enter Motif Pattern to search (Pattern containing only A, C, G, T): ").strip().upper()

    print(f"\nEvaluating: T = {text}, P = {pattern}")
    run_test(text, pattern)

if __name__ == "__main__":
    interactive_dna_test()
```

# LAB 4 : HASHING

## Problem

RailSwift is a start-up that runs ticket-booking counters at a busy railway junction. During festival season, counter agents must retrieve a passenger's booking within milliseconds using the 10-digit PNR (Passenger Name Record) number. The current system stores bookings in an unsorted list and performs a linear search, so lookups slow down sharply as bookings grow, and queues at the counter are getting longer.

You have been hired as a junior engineer to build the core lookup engine as a hash table.

Each booking record contains:

• PNR — 10-digit integer key (e.g. 8203416572); must be unique

• Passenger Name — string

• Train Number — 5-digit integer

• Status — CONFIRMED / RAC / WAITLIST

Because the server has limited memory, the engineering lead wants the table to start small and grow only when necessary. Cancelled tickets must be removed without breaking the lookup of
other bookings.

## Approach

### 1. Open Addressing Design
- **Hash Table Representation**: An array of slots is used, starting with an initial size of $m = 11$.
- **Slot States**: Each slot can be in one of three states:
  - `EMPTY` ($0$)
  - `OCCUPIED` ($1$)
  - `DELETED` ($2$, tombstone)
- **Hash Function**: Division method, defined as:
  $$h(k) = k \pmod m$$
- **Collision Resolution**: Linear probing, defined as:
  $$h(k, i) = (h(k) + i) \pmod m \quad \text{for } i \ge 0$$

### 2. Tombstone Semantics
- **Deletion**: Deleting a record updates the slot status to `DELETED` (tombstone) rather than resetting it to `EMPTY`. If it were reset to `EMPTY`, subsequent lookups for keys that collided at the same home index would terminate prematurely, resulting in false "Not Found" errors.
- **Insertion**: When inserting, if a tombstone is encountered, probing must still continue until an `EMPTY` slot is reached to verify that the key does not already exist in the cluster (preventing duplicate insertion). Once verified as a non-duplicate, the record is placed into the first encountered tombstone slot.

### 3. Dynamic Rehashing
- **Load Factor**: Defined as $\alpha = \frac{n}{m}$, where $n$ is the number of `OCCUPIED` slots.
- **Trigger**: If $\alpha > 0.7$ after an insertion, the table undergoes dynamic resizing.
- **Resizing**: The table size is increased to the smallest prime number $m_{\text{new}} \ge 2m$. Every active `OCCUPIED` record is rehashed using the new modulus, and all `DELETED` tombstones are purged.

### 4. Data Types
- A 10-digit PNR (e.g., $8203416572$) exceeds the standard 32-bit signed integer limit ($2^{31} - 1 \approx 2.14 \times 10^9$). Python handles arbitrarily large integers natively, but in platforms like C/C++, a 64-bit integer (`long long`) must be explicitly used to prevent overflow during calculations.

```python
EMPTY = 0
OCCUPIED = 1
DELETED = 2


def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True


def next_prime(num):
    while not is_prime(num):
        num += 1
    return num


def validate_pnr(pnr):
    return isinstance(pnr, int) and 1000000000 <= pnr <= 9999999999


class RailSwiftHashTable:
    def __init__(self, size=11, enable_rehashing=True):
        self.m = size
        self.enable_rehashing = enable_rehashing
        self.table = [None] * self.m
        self.state = [EMPTY] * self.m
        self.occupied_count = 0

    def hash(self, pnr):
        return pnr % self.m

    def load_factor(self):
        return self.occupied_count / self.m

    def display(self):
        print(f"\n--- Hash Table State (Size = {self.m}, Occupied = {self.occupied_count}, alpha = {self.load_factor():.3f}) ---")
        for i in range(self.m):
            if self.state[i] == OCCUPIED:
                rec = self.table[i]
                print(f"Slot {i:2d}: [OCCUPIED] PNR: {rec['pnr']} | {rec['name']} | Train: {rec['train_no']} | {rec['status']}")
            elif self.state[i] == DELETED:
                print(f"Slot {i:2d}: [DELETED / TOMBSTONE]")
            else:
                print(f"Slot {i:2d}: [EMPTY]")
        print("-" * 75)

    def insert(self, record):
        pnr = record.get("pnr")
        if not validate_pnr(pnr):
            print(f"[Error] Invalid PNR '{pnr}'. PNR must be a positive 10-digit numeric integer.")
            return False

        home_idx = self.hash(pnr)
        first_tombstone_idx = -1
        target_idx = -1
        probes = 0

        # Probe up to m times to avoid infinite loop
        for i in range(self.m):
            idx = (home_idx + i) % self.m
            probes += 1

            if self.state[idx] == OCCUPIED:
                if self.table[idx]["pnr"] == pnr:
                    print(f"[Error] Duplicate PNR: {pnr} already exists at index {idx}.")
                    return False
            elif self.state[idx] == DELETED:
                if first_tombstone_idx == -1:
                    first_tombstone_idx = idx
            elif self.state[idx] == EMPTY:
                target_idx = idx
                break

        # Check for overflow when table is full
        if target_idx == -1 and first_tombstone_idx == -1:
            print("[Error] Hash table overflow: Table is full.")
            return False

        # Reuse tombstone slot if available, else insert into empty slot
        final_idx = first_tombstone_idx if first_tombstone_idx != -1 else target_idx
        self.table[final_idx] = record
        self.state[final_idx] = OCCUPIED
        self.occupied_count += 1

        print(f"Inserted PNR {pnr} -> Home: {home_idx}, Final: {final_idx}, Probes: {probes}")

        # Check load factor and rehash if threshold is exceeded
        if self.enable_rehashing and self.load_factor() > 0.7:
            self.rehash()

        return True

    def search(self, pnr):
        if not validate_pnr(pnr):
            print(f"[Error] Invalid PNR '{pnr}'.")
            return None, 0

        if self.occupied_count == 0:
            print("[Notice] Table is empty.")
            return None, 0

        home_idx = self.hash(pnr)
        probes = 0

        for i in range(self.m):
            idx = (home_idx + i) % self.m
            probes += 1

            if self.state[idx] == EMPTY:
                print(f"Search PNR {pnr} -> Not Found (Stopped at EMPTY slot {idx}, Probes: {probes})")
                return None, probes
            elif self.state[idx] == OCCUPIED and self.table[idx]["pnr"] == pnr:
                rec = self.table[idx]
                print(f"Search PNR {pnr} -> Found at Slot {idx}! ({rec['name']}, {rec['status']}, Probes: {probes})")
                return rec, probes

        print(f"Search PNR {pnr} -> Not Found (Full table probed, Probes: {probes})")
        return None, probes

    def delete(self, pnr):
        if not validate_pnr(pnr):
            print(f"[Error] Invalid PNR '{pnr}'.")
            return False

        if self.occupied_count == 0:
            print("[Notice] Table is empty.")
            return False

        home_idx = self.hash(pnr)
        for i in range(self.m):
            idx = (home_idx + i) % self.m
            if self.state[idx] == EMPTY:
                break
            elif self.state[idx] == OCCUPIED and self.table[idx]["pnr"] == pnr:
                # Mark as DELETED instead of EMPTY to prevent breaking search chains
                self.state[idx] = DELETED
                self.table[idx] = None
                self.occupied_count -= 1
                print(f"Deleted PNR {pnr} at slot {idx}. Slot marked as [DELETED / TOMBSTONE].")
                return True

        print(f"Delete Failed: PNR {pnr} not found.")
        return False

    def rehash(self):
        new_size = next_prime(2 * self.m)
        print(f"\n[!] Load factor {self.load_factor():.3f} exceeded 0.70! Rehashing from size {self.m} to {new_size}...")

        old_table = self.table
        old_state = self.state
        old_m = self.m

        self.m = new_size
        self.table = [None] * self.m
        self.state = [EMPTY] * self.m
        self.occupied_count = 0

        for i in range(old_m):
            if old_state[i] == OCCUPIED:
                # Directly insert into the new table
                self.insert(old_table[i])
        print("[!] Rehashing completed successfully.\n")


# Execution of Mandatory Task 5 Demonstration Steps
def run_task5_demo():
    print("==================================================")
    print("      RAILSWIFT HASH TABLE DEMONSTRATION")
    print("==================================================")
    ht = RailSwiftHashTable(size=11, enable_rehashing=True)

    test_data = [
        {"pnr": 8203416572, "name": "Ananya Rao", "train_no": 12627, "status": "CONFIRMED"},
        {"pnr": 8203416583, "name": "Rahul Menon", "train_no": 12627, "status": "RAC"},
        {"pnr": 8203416540, "name": "Fatima Sheikh", "train_no": 16526, "status": "CONFIRMED"},
        {"pnr": 8203416599, "name": "Vikram Iyer", "train_no": 12008, "status": "WAITLIST"},
        {"pnr": 8203416615, "name": "Divya Nair", "train_no": 12627, "status": "CONFIRMED"},
        {"pnr": 8203416561, "name": "Arjun Patil", "train_no": 16526, "status": "RAC"},
        {"pnr": 8203416604, "name": "Sneha Kulkarni", "train_no": 12008, "status": "CONFIRMED"},
    ]

    print("\n--- Inserting Initial 7 Bookings ---")
    for rec in test_data:
        ht.insert(rec)

    print("\n--- Step 1: Display Table and Load Factor after Insertion #7 ---")
    ht.display()
    print(f"Current Load Factor: {ht.load_factor():.3f}")

    print("\n--- Step 2: Delete PNR 8203416583 (Rahul Menon) ---")
    ht.delete(8203416583)
    ht.display()

    print("\n--- Step 3: Search PNR 8203416561 (Arjun Patil) ---")
    ht.search(8203416561)

    print("\n--- Step 4: Search Deleted PNR 8203416583 ---")
    ht.search(8203416583)

    print("\n--- Step 5: Insert Record #8 (PNR 8203416626: Karthik Gowda) ---")
    record_8 = {"pnr": 8203416626, "name": "Karthik Gowda", "train_no": 12627, "status": "WAITLIST"}
    ht.insert(record_8)
    ht.display()

    print("\n--- Step 6: Demonstrating Boundary Conditions B1 - B7 ---")
    # B1: Duplicate PNR
    print("\n[B1 Check: Duplicate PNR]")
    ht.insert({"pnr": 8203416572, "name": "Duplicate User", "train_no": 12627, "status": "CONFIRMED"})

    # B2: Search/Delete on Empty Table
    print("\n[B2 Check: Empty Table]")
    empty_ht = RailSwiftHashTable(size=11)
    empty_ht.search(8203416572)
    empty_ht.delete(8203416572)

    # B3: Delete Non-Existent PNR
    print("\n[B3 Check: Delete Non-Existent PNR]")
    ht.delete(9999999999)

    # B4: Invalid PNR (Negative, non-10 digit)
    print("\n[B4 Check: Invalid PNR format]")
    ht.insert({"pnr": -12345, "name": "Invalid PNR", "train_no": 12345, "status": "RAC"})
    ht.insert({"pnr": 123, "name": "Short PNR", "train_no": 12345, "status": "RAC"})

    # B5: Search key beyond DELETED slot
    print("\n[B5 Check: Key beyond DELETED slot]")
    print("Verified in Step 3 where PNR 8203416561 was found across slot 0 (tombstone).")

    # B6: Insert reusing tombstone slot
    print("\n[B6 Check: Tombstone slot reuse]")
    print("Verified in Step 5 where PNR 8203416626 reused slot 0 after validating no duplicates.")

    # B7: Hash Table Overflow with Rehashing Disabled
    print("\n[B7 Check: Hash table overflow with rehashing off]")
    small_ht = RailSwiftHashTable(size=3, enable_rehashing=False)
    small_ht.insert({"pnr": 1000000001, "name": "User 1", "train_no": 11111, "status": "CONFIRMED"})
    small_ht.insert({"pnr": 1000000002, "name": "User 2", "train_no": 11111, "status": "CONFIRMED"})
    small_ht.insert({"pnr": 1000000003, "name": "User 3", "train_no": 11111, "status": "CONFIRMED"})
    small_ht.insert({"pnr": 1000000004, "name": "User 4", "train_no": 11111, "status": "CONFIRMED"})


if __name__ == "__main__":
    run_task5_demo()
```

# LAB 5: BREADTH FIRST SEARCH AND DEPTH FIRST SEARCH (BFS & DFS)

## Problem
Heavy rain has flooded parts of a city. The Disaster Management Authority runs a Control Headquarters (Zone A) and must reach several relief zones over a network of roads that are still usable. Every usable road connects two zones and can be traveled in both directions. Some zones have been completely cut off by collapsed bridges.

You are the software engineer on the response team. The authority needs a tool that will:
* **Plan dispatch in waves** — first the zones one road away from HQ, then two roads away, and so on (Breadth-First Search approach).
* **Find shortest paths** — identify the route with the fewest roads from HQ to any zone requesting urgent help.
* **Explore deeply** — explore the network deeply from HQ, simulating how a single rescue boat would follow one route as far as it goes before backtracking (Depth-First Search approach).
* **Report disconnected zones** — report which zones cannot be reached at all, so that air support/helicopters can be dispatched there instead.



## Approach

### 1. Graph Representation (Adjacency List)
* **Dynamic Mapping**: Vertices representing the geographic zones (e.g., zones $A$ through $J$) are mapped dynamically to array indices.
* **Undirected Symmetrical Edges**: Because roads are two-way, every edge $u-v$ is inserted symmetrically: $v$ is added to $u$'s adjacency list, and $u$ is added to $v$'s list.
* **Alphabetical Sorting**: Neighbors in each adjacency list are sorted alphabetically to guarantee consistent and reproducible BFS and DFS traversal orders.

### 2. Task 2: BFS Dispatch Waves
* **FIFO Queue**: Uses a First-In-First-Out (FIFO) queue starting at the Headquarters ($HQ$).
* **State Tracking**: Tracks visited nodes, level distances (representing dispatch waves), and parent pointers.
* **Level-by-Level Traversal**: Level 0 corresponds to $HQ$, while each subsequent wave processes vertices at distance $d+1$.

### 3. Task 3: Fewest-Roads Route
* **Backtracking**: Traverses backwards from the destination target zone $T$ to $HQ$ using the parent pointers recorded during the BFS execution.
* **Reversal**: Reversing this backtracked path yields the absolute shortest route (by number of edges) in an unweighted graph.

### 4. Task 4: DFS Rescue-Boat Exploration
* **Recursive Stack**: Implements a standard Depth-First Search via recursion (simulating an explicit stack).
* **Order of Traversal**: Visits neighbors alphabetically to systematically trace paths as deeply as possible before backtracking.

### 5. Task 5: Cut-off Zones & Connected Components
* **Multi-component Traversal**: Restarts the DFS from every unvisited node across the entire vertex set of the graph.
* **Unreachable Identification**: Any connected component that does not contain $HQ$ is flagged as completely unreachable, indicating zones where helicopters/air support must be deployed.


```python
from collections import deque


def solve():
    # ------------------ Input Handling ------------------
    # Line 1: V E
    try:
        print("Enter number of Vertices (V) and Edges (E) separated by a space (e.g., '10 12'):")
        header = input().strip()
        while not header:
            header = input().strip()
        V_count, E_count = map(int, header.split())
    except EOFError:
        return

    edges = []
    print(f"Enter the {E_count} edges (each edge as two space-separated zones, e.g., 'A B'):")
    for i in range(E_count):
        u, v = input().strip().split()
        edges.append((u, v))

    print("Enter the Source HQ and Target Zone separated by a space (e.g., 'A J'):")
    source_hq, target_zone = input().strip().split()

    # Collect all unique zones and sort them alphabetically
    all_zones = set([source_hq, target_zone])
    for u, v in edges:
        all_zones.add(u)
        all_zones.add(v)
    zones = sorted(list(all_zones))

    # Task 1: Build & Print Adjacency List
    adj = {z: [] for z in zones}
    for u, v in edges:
        adj[u].append(v)
        adj[v].append(u)

    for z in zones:
        adj[z].sort()

    print("\n=== TASK 1: Adjacency List ===")
    for z in zones:
        print(f"{z} -> {', '.join(adj[z]) if adj[z] else 'None'}")

    # ------------------ Task 2: BFS Dispatch Waves ------------------
    visited_bfs = {z: False for z in zones}
    distance = {z: -1 for z in zones}
    parent = {z: None for z in zones}
    bfs_order = []

    queue = deque([source_hq])
    visited_bfs[source_hq] = True
    distance[source_hq] = 0

    while queue:
        curr = queue.popleft()
        bfs_order.append(curr)

        for nbr in adj[curr]:
            if not visited_bfs[nbr]:
                visited_bfs[nbr] = True
                distance[nbr] = distance[curr] + 1
                parent[nbr] = curr
                queue.append(nbr)

    print("\n=== TASK 2: BFS Dispatch Waves ===")
    print(f"(a) BFS Visiting Order from HQ: {' -> '.join(bfs_order)}")
    print("(b) Levels (Roads from HQ):")
    for z in zones:
        if distance[z] != -1:
            print(f"    Zone {z}: {distance[z]} road(s)")
        else:
            print(f"    Zone {z}: Unreachable")

    # Group by waves
    max_dist = max(distance.values())
    print("(c) Zones Grouped by Wave:")
    for d in range(max_dist + 1):
        wave_nodes = [z for z in zones if distance[z] == d]
        if wave_nodes:
            print(f"    Wave {d} ({d} road(s) away): {', '.join(wave_nodes)}")

    # ------------------ Task 3: Fewest-Roads Route ------------------
    print(f"\n=== TASK 3: Fewest-Roads Route to {target_zone} ===")
    if not visited_bfs[target_zone]:
        print(f"Target zone {target_zone} cannot be reached from HQ by road!")
    else:
        path = []
        curr = target_zone
        while curr is not None:
            path.append(curr)
            curr = parent[curr]
        path.reverse()
        print(f"Route: {' -> '.join(path)}")
        print(f"Route Length: {len(path) - 1} road(s)")

    # ------------------ Task 4: DFS Rescue-Boat Exploration ------------------
    visited_dfs_hq = {z: False for z in zones}
    dfs_order = []

    def dfs_hq(node):
        visited_dfs_hq[node] = True
        dfs_order.append(node)
        for nbr in adj[node]:
            if not visited_dfs_hq[nbr]:
                dfs_hq(nbr)

    dfs_hq(source_hq)
    print("\n=== TASK 4: DFS Rescue-Boat Visiting Order ===")
    print(" -> ".join(dfs_order))

    # ------------------ Task 5: Connected Components & Cut-off Zones ------------------
    visited_components = {z: False for z in zones}
    components = []

    def dfs_comp(node, comp_list):
        visited_components[node] = True
        comp_list.append(node)
        for nbr in adj[node]:
            if not visited_components[nbr]:
                dfs_comp(nbr, comp_list)

    for z in zones:
        if not visited_components[z]:
            comp = []
            dfs_comp(z, comp)
            components.append(sorted(comp))

    print("\n=== TASK 5: Connected Components & Cut-Off Zones ===")
    print(f"Total Connected Components: {len(components)}")
    unreachable_zones = []
    for idx, comp in enumerate(components, 1):
        has_hq = source_hq in comp
        tag = " (Reachable from HQ)" if has_hq else " (Cut-Off: Send Helicopters!)"
        print(f"Component {idx}: {', '.join(comp)}{tag}")
        if not has_hq:
            unreachable_zones.extend(comp)

    print(f"Zones completely cut off from HQ: {', '.join(sorted(unreachable_zones)) if unreachable_zones else 'None'}")

    # ------------------ Self-Learning 1: Cycle / Loop Detection ------------------
    print("\n=== SELF-LEARNING: Loop / Cycle Detection ===")
    cycle_visited = {z: False for z in zones}
    cycle_parent = {z: None for z in zones}
    cycle_path = []

    def find_cycle(u, p):
        cycle_visited[u] = True
        cycle_parent[u] = p
        for v in adj[u]:
            if not cycle_visited[v]:
                if find_cycle(v, u):
                    return True
            elif v != p:
                # Cycle found! Backtrack from u to v
                curr = u
                cycle_path.append(v)
                while curr != v and curr is not None:
                    cycle_path.append(curr)
                    curr = cycle_parent[curr]
                cycle_path.append(v)
                return True
        return False

    has_cycle = False
    for z in zones:
        if not cycle_visited[z]:
            if find_cycle(z, None):
                has_cycle = True
                break

    if has_cycle:
        cycle_path.reverse()
        print("Cycle Detected: YES")
        print(f"Sample Loop Route: {' -> '.join(cycle_path)}")
    else:
        print("Cycle Detected: NO")


if __name__ == "__main__":
    solve()
```

## Time and Space Complexity

**Time Complexity:**

Both BFS and DFS run in $\mathcal{O}(V + E)$. Every vertex is visited at most once, and every edge is examined twice (once from each endpoint) in the adjacency list.   

**Space Complexity:**

$\mathcal{O}(V)$ auxiliary space. BFS requires a FIFO queue storing up to $V$ vertices alongside arrays for visited, distance, and parent. DFS uses up to $\mathcal{O}(V)$ stack space (call-stack recursion depth in the worst case of a linear graph).   

# LAB 6: SELF BALANCING BINARY SEARCH TREE : IMPLEMENTATION AND COMPARISION

## Approach

**Problem Context:**

In the MediQueue telemedicine system, patient IDs arrive from an auto-increment counter (sequential order).   

A standard Binary Search Tree (BST) degenerates into a skewed singly linked list of height $\mathcal{O}(N)$ under sorted insertions, causing search time to degrade from $\mathcal{O}(\log N)$ to $\mathcal{O}(N)$.   

An AVL Tree enforces the balance invariant: for every node, the heights of its left and right subtrees differ by at most 1 ($\text{Balance Factor} \in \{-1, 0, +1\}$). This guarantees an upper bound on height of $\approx 1.44 \log_2(N + 2)$, preserving strictly logarithmic $\mathcal{O}(\log N)$ operations.   

**Balancing Mechanics (Rotations):**

Balance Factor is defined as: $\text{BF}(node) = \text{height}(node.left) - \text{height}(node.right)$.   

Left-Left (LL): $\text{BF} > 1$ and key inserted into left child's left subtree $\implies$ Single Right Rotation at node.   

Right-Right (RR): $\text{BF} < -1$ and key inserted into right child's right subtree $\implies$ Single Left Rotation at node.   

Left-Right (LR): $\text{BF} > 1$ and key inserted into left child's right subtree $\implies$ Left Rotation on left child, then Right Rotation on node.   

Right-Left (RL): $\text{BF} < -1$ and key inserted into right child's left subtree $\implies$ Right Rotation on right child, then Left Rotation on node.   

**Handling Recursion Limits:**

For sequential datasets with $N = 10{,}000$, standard recursion in Python exceeds the default recursion depth ($1{,}000$). The Plain BST uses an iterative insertion algorithm to avoid call-stack overflow.   

## Implementation and demonstration

```python
import sys
import time
import random

# Ensure recursion limit handles AVL recursions if needed
sys.setrecursionlimit(20000)

# =====================================================================
# PART A: PLAIN BINARY SEARCH TREE (Iterative insert for N=10,000)
# =====================================================================
class BSTNode:
    def __init__(self, key, name):
        self.key = key
        self.name = name
        self.left = None
        self.right = None

class PlainBST:
    def __init__(self):
        self.root = None

    def insert(self, key, name="Patient"):
        new_node = BSTNode(key, name)
        if self.root is None:
            self.root = new_node
            return

        # Iterative insertion to prevent recursion stack overflow
        curr = self.root
        while True:
            if key < curr.key:
                if curr.left is None:
                    curr.left = new_node
                    break
                curr = curr.left
            elif key > curr.key:
                if curr.right is None:
                    curr.right = new_node
                    break
                curr = curr.right
            else:
                return  # Duplicate key ignored

    def search(self, key):
        curr = self.root
        comparisons = 0
        while curr is not None:
            comparisons += 1
            if key == curr.key:
                return curr, comparisons
            elif key < curr.key:
                curr = curr.left
            else:
                curr = curr.right
        return None, comparisons

    def delete(self, key):
        def _delete(node, k):
            if node is None:
                return None
            if k < node.key:
                node.left = _delete(node.left, k)
            elif k > node.key:
                node.right = _delete(node.right, k)
            else:
                # Case 1: Leaf node
                if node.left is None and node.right is None:
                    return None
                # Case 2: One child
                if node.left is None:
                    return node.right
                if node.right is None:
                    return node.left
                # Case 3: Two children (Find Inorder Successor)
                successor = node.right
                while successor.left is not None:
                    successor = successor.left
                node.key = successor.key
                node.name = successor.name
                node.right = _delete(node.right, successor.key)
            return node

        self.root = _delete(self.root, key)

    def height(self):
        def _get_height(node):
            if node is None:
                return 0
            return 1 + max(_get_height(node.left), _get_height(node.right))
        return _get_height(self.root)

    def inorder(self):
        res = []
        def _inorder(node):
            if node:
                _inorder(node.left)
                res.append(node.key)
                _inorder(node.right)
        _inorder(self.root)
        return res


# =====================================================================
# PART B: SELF-BALANCING AVL TREE
# =====================================================================
class AVLNode:
    def __init__(self, key, name):
        self.key = key
        self.name = name
        self.left = None
        self.right = None
        self.height = 1

class AVLTree:
    def __init__(self):
        self.root = None
        self.rotation_count = 0

    def _height(self, node):
        return node.height if node else 0

    def _balance_factor(self, node):
        return self._height(node.left) - self._height(node.right) if node else 0

    def _right_rotate(self, y):
        x = y.left
        T2 = x.right

        # Perform rotation
        x.right = y
        y.left = T2

        # Update heights
        y.height = 1 + max(self._height(y.left), self._height(y.right))
        x.height = 1 + max(self._height(x.left), self._height(x.right))

        self.rotation_count += 1
        return x

    def _left_rotate(self, x):
        y = x.right
        T2 = y.left

        # Perform rotation
        y.left = x
        x.right = T2

        # Update heights
        x.height = 1 + max(self._height(x.left), self._height(x.right))
        y.height = 1 + max(self._height(y.left), self._height(y.right))

        self.rotation_count += 1
        return y

    def insert(self, key, name="Patient"):
        def _insert(node, k, n):
            if node is None:
                return AVLNode(k, n)

            if k < node.key:
                node.left = _insert(node.left, k, n)
            elif k > node.key:
                node.right = _insert(node.right, k, n)
            else:
                return node  # Duplicate key

            # Update height of ancestor node
            node.height = 1 + max(self._height(node.left), self._height(node.right))
            bf = self._balance_factor(node)

            # Rebalancing cases
            # Case 1: Left-Left (LL)
            if bf > 1 and k < node.left.key:
                return self._right_rotate(node)

            # Case 2: Right-Right (RR)
            if bf < -1 and k > node.right.key:
                return self._left_rotate(node)

            # Case 3: Left-Right (LR)
            if bf > 1 and k > node.left.key:
                node.left = self._left_rotate(node.left)
                return self._right_rotate(node)

            # Case 4: Right-Left (RL)
            if bf < -1 and k < node.right.key:
                node.right = self._right_rotate(node.right)
                return self._left_rotate(node)

            return node

        self.root = _insert(self.root, key, name)

    def delete(self, key):
        def _delete(node, k):
            if node is None:
                return None

            if k < node.key:
                node.left = _delete(node.left, k)
            elif k > node.key:
                node.right = _delete(node.right, k)
            else:
                if node.left is None or node.right is None:
                    node = node.left if node.left else node.right
                else:
                    successor = node.right
                    while successor.left is not None:
                        successor = successor.left
                    node.key = successor.key
                    node.name = successor.name
                    node.right = _delete(node.right, successor.key)

            if node is None:
                return None

            node.height = 1 + max(self._height(node.left), self._height(node.right))
            bf = self._balance_factor(node)

            # Rebalance on delete
            if bf > 1 and self._balance_factor(node.left) >= 0:
                return self._right_rotate(node)
            if bf > 1 and self._balance_factor(node.left) < 0:
                node.left = self._left_rotate(node.left)
                return self._right_rotate(node)
            if bf < -1 and self._balance_factor(node.right) <= 0:
                return self._left_rotate(node)
            if bf < -1 and self._balance_factor(node.right) > 0:
                node.right = self._right_rotate(node.right)
                return self._left_rotate(node)

            return node

        self.root = _delete(self.root, key)

    def search(self, key):
        curr = self.root
        comparisons = 0
        while curr is not None:
            comparisons += 1
            if key == curr.key:
                return curr, comparisons
            elif key < curr.key:
                curr = curr.left
            else:
                curr = curr.right
        return None, comparisons

    def height(self):
        return self._height(self.root)

    def inorder(self):
        res = []
        def _inorder(node):
            if node:
                _inorder(node.left)
                res.append(node.key)
                _inorder(node.right)
        _inorder(self.root)
        return res

    def is_balanced(self):
        def _check(node):
            if node is None:
                return True
            if abs(self._balance_factor(node)) > 1:
                return False
            return _check(node.left) and _check(node.right)
        return _check(self.root)

    def is_bst(self):
        inorder_keys = self.inorder()
        return all(inorder_keys[i] < inorder_keys[i + 1] for i in range(len(inorder_keys) - 1))


# =====================================================================
# SECTION 3 & WORKLOAD BENCHMARK RUNNER
# =====================================================================
def run_sample_scenario():
    print("==================================================")
    print("   SECTION 3: SAMPLE INPUT / OUTPUT VERIFICATION   ")
    print("==================================================")
    sample_ids = [10, 20, 30, 40, 50, 25]

    bst = PlainBST()
    avl = AVLTree()

    for pid in sample_ids:
        bst.insert(pid)
        avl.insert(pid)

    # Verification assertions
    assert avl.is_balanced(), "AVL Tree balance check failed!"
    assert avl.is_bst(), "AVL Tree BST property check failed!"

    print(f"{'Observation':<26} | {'Plain BST':<12} | {'AVL Tree'}")
    print("-" * 65)
    print(f"{'Root after all insertions':<26} | {bst.root.key:<12} | {avl.root.key}")
    print(f"{'Height':<26} | {bst.height():<12} | {avl.height()}")
    print(f"{'Rotations performed':<26} | {'0':<12} | {avl.rotation_count} single rotations")
    print(f"{'In-order traversal':<26} | {' '.join(map(str, bst.inorder())):<12} | {' '.join(map(str, avl.inorder()))}")
    print("=" * 65)


if __name__ == "__main__":
    run_sample_scenario()
```

## Self Learning Component: Red-Black Trees

### 1. Core Balancing Idea & Color Rules
A Red-Black Tree (RBT) is a self-balancing binary search tree where each node stores an extra color bit (**RED** or **BLACK**). It maintains balance through five strict structural invariants:
1. Every node is either **RED** or **BLACK**.
2. The root is always **BLACK**.
3. Every leaf (`NIL` sentinel) is **BLACK**.
4. If a node is **RED**, both its children must be **BLACK** (no two consecutive **RED** nodes allowed).
5. For each node, every simple path from that node to any of its descendant `NIL` leaves contains the exact same number of black nodes (known as the *black-height*).

### 2. Height Bound: AVL vs. Red-Black Tree
* **AVL Bound:** $\text{Height} \le 1.44 \log_2(N + 2)$ due to its rigid balance factor constraint (left and right subtree heights differ by at most 1).
* **Red-Black Bound:** $\text{Height} \le 2 \log_2(N + 1)$.

### 3. Practical Differences & Trade-offs
* **Search Operations:** AVL trees are more rigidly balanced than Red-Black trees. Consequently, AVL trees guarantee slightly faster lookup and search performance because of their shorter structural height.
* **Insertion & Deletion:** Red-Black trees require at most $2$ rotations per insertion and $3$ rotations per deletion, relying mostly on inexpensive color flips. As a result, RBTs outperform AVL trees on insert/delete-heavy workloads.
* **Industry Standard:** Standard libraries (such as C++ `std::map`, Java `TreeMap`, and the Linux kernel Completely Fair Scheduler) standardize on **Red-Black Trees** due to this balanced write-heavy performance.

### 4. Simple Red-Black Tree Implementation
Below is a python implementation demonstrating node insertion, rebalancing (via rotations and color flips), and verification of Red-Black Tree properties.

```python
RED = True
BLACK = False

class RBTNode:
    def __init__(self, key, color=RED):
        self.key = key
        self.color = color
        self.left = None
        self.right = None
        self.parent = None

class RedBlackTree:
    def __init__(self):
        # Use a single sentinel node for all NIL leaves
        self.NIL = RBTNode(0, color=BLACK)
        self.root = self.NIL

    def left_rotate(self, x):
        y = x.right
        x.right = y.left
        if y.left != self.NIL:
            y.left.parent = x
        y.parent = x.parent
        if x.parent is None:
            self.root = y
        elif x == x.parent.left:
            x.parent.left = y
        else:
            x.parent.right = y
        y.left = x
        x.parent = y

    def right_rotate(self, y):
        x = y.left
        y.left = x.right
        if x.right != self.NIL:
            x.right.parent = y
        x.parent = y.parent
        if y.parent is None:
            self.root = x
        elif y == y.parent.left:
            y.parent.left = x
        else:
            y.parent.right = x
        x.right = y
        y.parent = x

    def insert(self, key):
        # Step 1: Standard BST Insertion with color = RED
        new_node = RBTNode(key, color=RED)
        new_node.left = self.NIL
        new_node.right = self.NIL

        y = None
        x = self.root
        while x != self.NIL:
            y = x
            if new_node.key < x.key:
                x = x.left
            else:
                x = x.right

        new_node.parent = y
        if y is None:
            self.root = new_node
        elif new_node.key < y.key:
            y.left = new_node
        else:
            y.right = new_node

        # Step 2: Fix Violations
        self._insert_fixup(new_node)

    def _insert_fixup(self, k):
        while k.parent and k.parent.color == RED:
            if k.parent == k.parent.parent.left:
                uncle = k.parent.parent.right
                if uncle.color == RED: # Case 1: Recolor
                    k.parent.color = BLACK
                    uncle.color = BLACK
                    k.parent.parent.color = RED
                    k = k.parent.parent
                else:
                    if k == k.parent.right: # Case 2: LR rotation
                        k = k.parent
                        self.left_rotate(k)
                    # Case 3: LL rotation
                    k.parent.color = BLACK
                    k.parent.parent.color = RED
                    self.right_rotate(k.parent.parent)
            else:
                uncle = k.parent.parent.left
                if uncle.color == RED: # Case 1: Recolor
                    k.parent.color = BLACK
                    uncle.color = BLACK
                    k.parent.parent.color = RED
                    k = k.parent.parent
                else:
                    if k == k.parent.left: # Case 2: RL rotation
                        k = k.parent
                        self.right_rotate(k)
                    # Case 3: RR rotation
                    k.parent.color = BLACK
                    k.parent.parent.color = RED
                    self.left_rotate(k.parent.parent)
            if k == self.root:
                break
        self.root.color = BLACK

    def inorder(self):
        keys = []
        def _inorder(node):
            if node != self.NIL:
                _inorder(node.left)
                keys.append(f"{node.key}({'R' if node.color == RED else 'B'})")
                _inorder(node.right)
        _inorder(self.root)
        return keys

# Quick Demo
if __name__ == "__main__":
    rbt = RedBlackTree()
    elements = [10, 20, 30, 15, 25, 5]
    print(f"Inserting sequence: {elements}")
    for el in elements:
        rbt.insert(el)

    print("\nIn-order Traversal (Key and Color):")
    print(" -> ".join(rbt.inorder()))
    print(f"Root of the RBT: {rbt.root.key} ({'RED' if rbt.root.color == RED else 'BLACK'})")
```

# LAB 7: GREEDY & DIVIDE AND CONQUER - DIJKSTRA'S ALGORITHM
(SET-A)

## Approach

### 1. Greedy Choice & Relaxation
Dijkstra's algorithm maintains a set of visited vertices whose shortest distances from the source are finalized. At each step, it greedily selects the unvisited vertex $u$ with the minimum tentative distance and relaxes all its outgoing edges $(u, v)$ with weight $w$:

$$\text{if } \text{dist}[u] + w < \text{dist}[v] \implies \text{dist}[v] = \text{dist}[u] + w, \quad \text{parent}[v] = u$$

### 2. Route Reconstruction
A parent array tracks the predecessor of each node along the optimal path, allowing route reconstruction via backtracking from the destination to the source.

### 3. Rush-Hour Edge Blockage (Task 3)
When a road $(u, v)$ is blocked, the edge weight is set to $\infty$ (or removed from the adjacency list) and Dijkstra is re-executed to identify affected localities and latency deltas.

### 4. Boundary & Negative Weight Detection (Task 4)
Input weights are validated to ensure $w \ge 0$. Any vertex retaining $\text{dist}[v] = \infty$ after termination is reported as "INF / Not reachable".

```python
import sys
import heapq

LOCALITY_NAMES = {
    0: "Hebbal Hub",
    1: "Indiranagar",
    2: "Koramangala",
    3: "MG Road",
    4: "Jayanagar",
    5: "Electronic City"
}

def dijkstra(n, adj, source):
    dist = [float('inf')] * n
    parent = [-1] * n
    dist[source] = 0

    # Priority Queue storing tuples: (current_distance, node)
    pq = [(0, source)]

    while pq:
        d, u = heapq.heappop(pq)

        if d > dist[u]:
            continue

        for v, w in adj[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    return dist, parent

def reconstruct_path(parent, target):
    path = []
    curr = target
    while curr != -1:
        path.append(curr)
        curr = parent[curr]
    path.reverse()
    return path

def print_routing_table(n, dist, parent, source):
    print("\n" + "=" * 65)
    print(f"{'Locality':<24} | {'Time (min)':<12} | {'Route'}")
    print("-" * 65)
    for i in range(n):
        name_str = f"{i} {LOCALITY_NAMES.get(i, f'Zone {i}')}"
        if dist[i] == float('inf'):
            print(f"{name_str:<24} | {'INF':<12} | Not reachable")
        else:
            path = reconstruct_path(parent, i)
            route_str = " -> ".join(map(str, path))
            print(f"{name_str:<24} | {dist[i]:<12} | {route_str}")
    print("=" * 65)

def main():
    # --- Sample Input Loading ---
    raw_input_data = """6 8
0 1 4
0 2 8
1 2 2
1 3 5
2 4 10
3 4 2
3 5 6
4 5 3
0"""

    lines = [line.strip() for line in raw_input_data.strip().split("\n") if line.strip()]
    n, m = map(int, lines[0].split())

    edges = []
    adj = {i: [] for i in range(n)}

    for k in range(1, m + 1):
        u, v, w = map(int, lines[k].split())
        # Task 4: Negative weight check
        if w < 0:
            print(f"[Error] Negative edge weight detected on road ({u}, {v}): {w}. Dijkstra aborted.")
            return
        edges.append((u, v, w))
        adj[u].append((v, w))
        adj[v].append((u, w))

    source = int(lines[m + 1])

    print("=== QUICKKART LAST-MILE DISPATCH ROUTING (BASELINE) ===")
    base_dist, base_parent = dijkstra(n, adj, source)
    print_routing_table(n, base_dist, base_parent, source)

    # --- Task 3: Rush-Hour Road Blockage Simulation ---
    print("\n=== TASK 3: RUSH HOUR ROAD BLOCKAGE SIMULATION ===")
    block_u, block_v = 3, 4  # Road between MG Road and Jayanagar blocked
    print(f"Simulating blocked road between Node {block_u} ({LOCALITY_NAMES[block_u]}) and Node {block_v} ({LOCALITY_NAMES[block_v]})...")

    # Rebuild adjacency list without blocked road
    blocked_adj = {i: [] for i in range(n)}
    for u, v, w in edges:
        if (u == block_u and v == block_v) or (u == block_v and v == block_u):
            continue
        blocked_adj[u].append((v, w))
        blocked_adj[v].append((u, w))

    new_dist, new_parent = dijkstra(n, blocked_adj, source)
    print_routing_table(n, new_dist, new_parent, source)

    print("\n--- Impact Summary ---")
    changes = 0
    for i in range(n):
        if new_dist[i] != base_dist[i]:
            changes += 1
            old_val = "INF" if base_dist[i] == float('inf') else f"{base_dist[i]} min"
            new_val = "INF" if new_dist[i] == float('inf') else f"{new_dist[i]} min"
            print(f"- {LOCALITY_NAMES[i]}: Travel time changed from {old_val} to {new_val} (Delta: +{new_dist[i] - base_dist[i]} min)")
    if changes == 0:
        print("No localities were impacted by this road blockage.")

if __name__ == "__main__":
    main()
```

# LAB 8: STRASSEN'S MATRIX MULTIPLICATION

## Problem

You have joined the performance engineering team of NeuraLens AI, a start-up that builds
image-classification models for rural health clinics. Their models run on low-cost servers,
and profiling shows that most of the training and inference time goes into multiplying large
dense weight matrices in the fully connected layers.


The team lead suspects that the classical O(n³) multiplication is the bottleneck and asks you
to build a prototype matrix multiplication module based on Strassen’s divide-and-conquer
algorithm. Before the company adopts it, you must prove that it is correct, show how it
behaves as the matrix size grows, and report honestly where it helps and where it does not.

## Approach

### 1. Classical Triple-Loop Baseline ($\mathcal{O}(n^3)$)
Computes each entry using three nested loops:
$$C[i][j] = \sum_{k=0}^{n-1} A[i][k] \cdot B[k][j]$$
This performs $n^3$ scalar multiplications and $n^2(n-1)$ additions. It serves as our ground-truth reference to verify correctness.

### 2. Strassen's Divide-and-Conquer Strategy ($\mathcal{O}(n^{\log_2 7}) \approx \mathcal{O}(n^{2.807})$)
- Divides two $n \times n$ matrices into four $n/2 \times n/2$ sub-matrices:
  - $A \rightarrow A_{11}, A_{12}, A_{21}, A_{22}$
  - $B \rightarrow B_{11}, B_{12}, B_{21}, B_{22}$
- Reduces the standard $8$ recursive block multiplications down to $7$ auxiliary formulas ($M_1$ through $M_7$) at the expense of $18$ matrix additions and subtractions.

### 3. Handling Arbitrary Dimensions (Zero-Padding & Trimming)
- Since real neural network weight matrices are rarely powers of two, we dynamically find the next power of two:
  $$m = 2^{\lceil \log_2(\max(r_A, c_A, r_B, c_B)) \rceil}$$
- We pad both matrices with zeros up to $m \times m$, execute Strassen's multiplication, and slice the final result back down to the target dimension $(r_A \times c_B)$.

### 4. Hybrid Cutoff Threshold ($T$)
- Recursing all the way down to $1 \times 1$ is inefficient due to high function-call overhead and matrix allocation penalties.
- When the sub-problem dimension $n \le T$, the recursion terminates and switches directly to the classical triple-loop method.

### 5. Verification Metric
- To account for tiny floating-point variations introduced by reordered operations, correctness is validated using an absolute tolerance:
  $$\max |C_{\text{strassen}}[i][j] - C_{\text{classical}}[i][j]| < 10^{-6}$$

```python
import time
import random
import math


class OpCounter:
    def __init__(self):
        self.muls = 0
        self.adds = 0


# =====================================================================
# MODULAR MATRIX HELPERS
# =====================================================================
def create_matrix(rows, cols, val=0.0):
    return [[val for _ in range(cols)] for _ in range(rows)]


def add_matrix(A, B, counter):
    n = len(A)
    C = create_matrix(n, n)
    for i in range(n):
        for j in range(n):
            C[i][j] = A[i][j] + B[i][j]
    counter.adds += n * n
    return C


def sub_matrix(A, B, counter):
    n = len(A)
    C = create_matrix(n, n)
    for i in range(n):
        for j in range(n):
            C[i][j] = A[i][j] - B[i][j]
    counter.adds += n * n
    return C


def split_matrix(P):
    n = len(P)
    half = n // 2
    P11 = [row[:half] for row in P[:half]]
    P12 = [row[half:] for row in P[:half]]
    P21 = [row[:half] for row in P[half:]]
    P22 = [row[half:] for row in P[half:]]
    return P11, P12, P21, P22


def combine_matrix(C11, C12, C21, C22):
    half = len(C11)
    n = half * 2
    C = create_matrix(n, n)
    for i in range(half):
        C[i][:half] = C11[i]
        C[i][half:] = C12[i]
        C[i + half][:half] = C21[i]
        C[i + half][half:] = C22[i]
    return C


# =====================================================================
# TASK 1: CLASSICAL BASELINE MULTIPLICATION
# =====================================================================
def classical_multiply(A, B, counter):
    rA, cA = len(A), len(A[0])
    cB = len(B[0])
    C = create_matrix(rA, cB, 0.0)

    for i in range(rA):
        row_a = A[i]
        row_c = C[i]
        for k in range(cA):
            a_ik = row_a[k]
            row_b = B[k]
            for j in range(cB):
                row_c[j] += a_ik * row_b[j]

    ops = rA * cA * cB
    counter.muls += ops
    counter.adds += ops
    return C


# =====================================================================
# TASK 2 & 4: RECURSIVE STRASSEN WITH CUTOFF THRESHOLD T
# =====================================================================
def strassen_recursive(A, B, T, counter):
    n = len(A)

    # Task 4: Base case cutoff to classical multiplication
    if n <= T:
        return classical_multiply(A, B, counter)

    # Split matrices into n/2 x n/2 quadrants
    A11, A12, A21, A22 = split_matrix(A)
    B11, B12, B21, B22 = split_matrix(B)

    # Compute the 7 Strassen products (M1 to M7)
    # M1 = (A11 + A22) * (B11 + B22)
    M1 = strassen_recursive(add_matrix(A11, A22, counter), add_matrix(B11, B22, counter), T, counter)

    # M2 = (A21 + A22) * B11
    M2 = strassen_recursive(add_matrix(A21, A22, counter), B11, T, counter)

    # M3 = A11 * (B12 - B22)
    M3 = strassen_recursive(A11, sub_matrix(B12, B22, counter), T, counter)

    # M4 = A22 * (B21 - B11)
    M4 = strassen_recursive(A22, sub_matrix(B21, B11, counter), T, counter)

    # M5 = (A11 + A12) * B22
    M5 = strassen_recursive(add_matrix(A11, A12, counter), B22, T, counter)

    # M6 = (A21 - A11) * (B11 + B12)
    M6 = strassen_recursive(sub_matrix(A21, A11, counter), add_matrix(B11, B12, counter), T, counter)

    # M7 = (A12 - A22) * (B21 + B22)
    M7 = strassen_recursive(sub_matrix(A12, A22, counter), add_matrix(B21, B22, counter), T, counter)

    # Combine products into result quadrants C11, C12, C21, C22
    # C11 = M1 + M4 - M5 + M7
    C11 = add_matrix(sub_matrix(add_matrix(M1, M4, counter), M5, counter), M7, counter)

    # C12 = M3 + M5
    C12 = add_matrix(M3, M5, counter)

    # C21 = M2 + M4
    C21 = add_matrix(M2, M4, counter)

    # C22 = M1 - M2 + M3 + M6
    C22 = add_matrix(add_matrix(sub_matrix(M1, M2, counter), M3, counter), M6, counter)

    return combine_matrix(C11, C12, C21, C22)


# =====================================================================
# TASK 3: ARBITRARY SIZE HANDLING (ZERO-PADDING & TRIMMING)
# =====================================================================
def next_power_of_two(x):
    return 1 if x <= 1 else 2 ** math.ceil(math.log2(x))


def strassen_multiply(A, B, T=32, counter=None):
    if counter is None:
        counter = OpCounter()

    rA, cA = len(A), len(A[0])
    rB, cB = len(B), len(B[0])
    if cA != rB:
        raise ValueError("Incompatible matrix dimensions for multiplication.")

    max_dim = max(rA, cA, rB, cB)
    padded_size = next_power_of_two(max_dim)

    # Pad A and B with zeros if needed
    if padded_size == rA == cA == rB == cB:
        A_pad, B_pad = A, B
    else:
        A_pad = create_matrix(padded_size, padded_size, 0.0)
        B_pad = create_matrix(padded_size, padded_size, 0.0)
        for i in range(rA):
            for j in range(cA):
                A_pad[i][j] = A[i][j]
        for i in range(rB):
            for j in range(cB):
                B_pad[i][j] = B[i][j]

    C_pad = strassen_recursive(A_pad, B_pad, T, counter)

    # Trim result back to original rA x cB dimensions
    return [row[:cB] for row in C_pad[:rA]]


# =====================================================================
# TASK 6: VALIDATION & BENCHMARK RUNNER
# =====================================================================
def validate_matrices(C_ref, C_test, tol=1e-6):
    for i in range(len(C_ref)):
        for j in range(len(C_ref[0])):
            if abs(C_ref[i][j] - C_test[i][j]) > tol:
                return False
    return True


def print_matrix(M, label):
    print(f"{label}:")
    for row in M:
        print("  [" + "  ".join(f"{val:6.2f}" if isinstance(val, float) else f"{val:4d}" for val in row) + "]")


def main():
    # 1. Sample Test (n = 2)
    print("==========================================================")
    print("  SAMPLE TEST (n = 2, Cutoff T = 1)")
    print("==========================================================")
    A_sample = [[1, 2], [3, 4]]
    B_sample = [[5, 6], [7, 8]]

    c_ref_ops = OpCounter()
    C_ref = classical_multiply(A_sample, B_sample, c_ref_ops)

    c_str_ops = OpCounter()
    C_str = strassen_multiply(A_sample, B_sample, T=1, counter=c_str_ops)

    print_matrix(A_sample, "Matrix A")
    print_matrix(B_sample, "Matrix B")
    print_matrix(C_str, "Product A x B (Strassen)")
    status = "PASS" if validate_matrices(C_ref, C_str) else "FAIL"
    print(f"Validation: {status}")
    print(f"Classical -> Muls: {c_ref_ops.muls}, Adds: {c_ref_ops.adds}")
    print(f"Strassen  -> Muls: {c_str_ops.muls}, Adds: {c_str_ops.adds}")

    # 2. Arbitrary Non-Power-of-2 Test (n = 3)
    print("\n==========================================================")
    print("  ARBITRARY SIZE TEST (n = 3 -> Padded to 4 -> Trimmed to 3)")
    print("==========================================================")
    A_3 = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
    B_3 = [[9, 8, 7], [6, 5, 4], [3, 2, 1]]
    C_3_ref = classical_multiply(A_3, B_3, OpCounter())
    C_3_str = strassen_multiply(A_3, B_3, T=2, counter=OpCounter())
    print_matrix(C_3_str, "Product A x B (3x3)")
    print(f"Validation: {'PASS' if validate_matrices(C_3_ref, C_3_str) else 'FAIL'}")

    # 3. Task 5 & 6 Benchmark (n = 2 to 512, Cutoff T = 32)
    print("\n=========================================================================================================")
    print(f"  BENCHMARK TABLE (n = 2 to 512, Cutoff Threshold T = 32)")
    print("=========================================================================================================")
    header = f"{'n':>4} | {'Class Muls':>11} | {'Stras Muls':>11} | {'Class Adds':>11} | {'Stras Adds':>11} | {'Class (ms)':>10} | {'Stras (ms)':>10} | {'Status'}"
    print(header)
    print("-" * len(header))

    random.seed(42)
    sizes = [2, 4, 8, 16, 32, 64, 128, 256, 512]
    results = []

    for n in sizes:
        A = [[random.uniform(-1.0, 1.0) for _ in range(n)] for _ in range(n)]
        B = [[random.uniform(-1.0, 1.0) for _ in range(n)] for _ in range(n)]

        ops_c = OpCounter()
        t0 = time.perf_counter()
        res_c = classical_multiply(A, B, ops_c)
        t_c = (time.perf_counter() - t0) * 1000.0

        ops_s = OpCounter()
        t0 = time.perf_counter()
        res_s = strassen_multiply(A, B, T=32, counter=ops_s)
        t_s = (time.perf_counter() - t0) * 1000.0

        valid = "PASS" if validate_matrices(res_c, res_s) else "FAIL"
        results.append((n, t_c, t_s))
        print(f"{n:4d} | {ops_c.muls:11d} | {ops_s.muls:11d} | {ops_c.adds:11d} | {ops_s.adds:11d} | {t_c:10.3f} | {t_s:10.3f} | {valid}")

    # ASCII Running Time Plot
    print("\n=== Running Time Comparison Plot (n vs Execution Time in ms) ===")
    max_t = max(max(tc, ts) for _, tc, ts in results)
    for n, tc, ts in results:
        bar_c = "#" * int((tc / max_t) * 40)
        bar_s = "*" * int((ts / max_t) * 40)
        print(f"n={n:3d} | Classical ({tc:7.2f} ms): {bar_c}")
        print(f"      | Strassen  ({ts:7.2f} ms): {bar_s}")


if __name__ == "__main__":
    main()
```

| **$T=512$ (Pure Classical)** | 134,217,728 | 134,217,728 | 0 levels | 8,910.4 ms | 1.00× (Baseline) |


### 2. Why the Optimal Cutoff $T$ is Not $1$

* **Addition Explosion at Small Sizes:** At $n = 2$, Strassen reduces multiplications from $8$ to $7$ (saving 1 multiplication), but increases additions from $8$ to $18$ (adding 10 extra additions). Recursing all the way to $T = 1$ balloons total additions to 286.9 million (more than double the classical total), wiping out the benefits of fewer multiplications.
* **Memory Allocation & Function Call Overhead:** Every recursive split allocates 18 temporary sub-matrices ($A_{11}\dots A_{22}$, $B_{11}\dots B_{22}$, $M_1\dots M_7$, $C_{11}\dots C_{22}$). At $T = 1$, there are $7^9 = 40,353,607$ leaf calls, causing massive heap/garbage-collection overhead.
* **CPU Cache Locality:** Once a sub-block shrinks to $T = 32$ or $T = 64$ ($32 \times 32 \times 8\text{ bytes} = 8\text{ KB}$), all three blocks ($A, B, C$) fit simultaneously inside the CPU's L1 Data Cache ($32\text{ KB}$–$64\text{ KB}$), allowing the classical triple-loop to run at peak hardware throughput without cache misses.

# LAB 9: LONGEST COMMON SUB SEQUENCE

## Problem

LexiSum AI is a start-up that builds an abstractive text summarisation model for legal and
news articles. Before every model release, the evaluation team must check how close the
machine-generated summary is to a human-written reference summary.
The team uses a metric called ROUGE-L, which is based on the Longest Common
Subsequence of words between the reference summary and the candidate (model)
summary. A subsequence keeps the order of words but need not be contiguous. For
example, &quot;the cat on the mat&quot; is a subsequence of &quot;the cat sat on the mat&quot;.
You have been hired as an intern in the evaluation team. Your manager has asked you to
build the core LCS engine from scratch (no library functions for LCS or sequence matching)
and use it to score model summaries.

Implement the Longest Common Subsequence algorithm using Dynamic Programming and
use it to compute word-level ROUGE-L scores for pairs of summaries.

## Approach

### 1. Text Preprocessing (Word-Level Tokenization)
Summary strings are converted to lowercase, stripped of punctuation, and split into word lists $R = [r_1, \dots, r_r]$ and $C = [c_1, \dots, c_c]$.   

### 2. Bottom-Up Dynamic Programming Table
Constructs a $(m + 1) \times (n + 1)$ table where $L[i][j]$ stores the LCS length of prefixes $X[0..i-1]$ and $Y[0..j-1]$:

$$L[i][j] = \begin{cases} 0 & \text{if } i = 0 \text{ or } j = 0 \\ L[i-1][j-1] + 1 & \text{if } X[i-1] = Y[j-1] \\ \max(L[i-1][j], L[i][j-1]) & \text{otherwise} \end{cases}$$

### 3. Sequence Reconstruction (Backtracking)
Starts at cell $(m, n)$ and traces backward to $(0, 0)$. When $X[i-1] = Y[j-1]$, the matching token is prepended to the result and indices move diagonally to $(i-1, j-1)$. Otherwise, traversal moves toward whichever neighbor ($L[i-1][j]$ or $L[i][j-1]$) holds the larger value.

### 4. Word-Level ROUGE-L Evaluation
Given reference length $r$, candidate length $c$, and LCS length $l$:

$$\text{Precision} = \frac{l}{c}, \qquad \text{Recall} = \frac{l}{r}, \qquad F_1 = \begin{cases} 0 & \text{if } l = 0 \\ \dfrac{2 \times \text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}} & \text{otherwise} \end{cases}$$

### 5. Edge-Case Handling
Empty summaries ($r = 0$ or $c = 0$), zero overlap ($l = 0$), identical texts ($F_1 = 1.000$), and repeated words are guarded against division-by-zero errors.

## Implementation and Demonstration

```python
import string


def preprocess_words(text):
    """Lowercase text, strip punctuation, and split into a list of words."""
    cleaned = "".join(ch.lower() if ch not in string.punctuation else " " for ch in text)
    return cleaned.split()


def compute_lcs_dp(X, Y):
    """Task 1: Build the (m+1) x (n+1) Dynamic Programming table for LCS."""
    m, n = len(X), len(Y)
    L = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if X[i - 1] == Y[j - 1]:
                L[i][j] = L[i - 1][j - 1] + 1
            else:
                L[i][j] = max(L[i - 1][j], L[i][j - 1])

    return L


def reconstruct_lcs(X, Y, L):
    """Task 2: Backtrack through DP table L to recover one actual LCS."""
    i, j = len(X), len(Y)
    lcs_tokens = []

    while i > 0 and j > 0:
        if X[i - 1] == Y[j - 1]:
            lcs_tokens.append(X[i - 1])
            i -= 1
            j -= 1
        elif L[i - 1][j] >= L[i][j - 1]:
            i -= 1
        else:
            j -= 1

    lcs_tokens.reverse()
    return lcs_tokens


def lcs_space_optimized(X, Y):
    """Self-Learning: Compute LCS length using O(min(m, n)) space (two rows)."""
    # Ensure Y is the shorter sequence so row width is min(m, n)
    if len(Y) > len(X):
        X, Y = Y, X

    m, n = len(X), len(Y)
    prev = [0] * (n + 1)
    curr = [0] * (n + 1)

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if X[i - 1] == Y[j - 1]:
                curr[j] = prev[j - 1] + 1
            else:
                curr[j] = max(prev[j], curr[j - 1])
        prev, curr = curr, [0] * (n + 1)

    return prev[n]


def print_dp_table(X, Y, L):
    """Print formatted DP table with sequence headers."""
    col_labels = ["0"] + [str(tok)[:6] for tok in Y]
    row_labels = ["0"] + [str(tok)[:6] for tok in X]

    print("DP Table:")
    header = f"{'':>7} | " + " ".join(f"{lbl:>6}" for lbl in col_labels)
    print(header)
    print("-" * len(header))

    for i, r_lbl in enumerate(row_labels):
        row_vals = " ".join(f"{val:6d}" for val in L[i])
        print(f"{r_lbl:>7} | {row_vals}")


def compute_rouge_l(ref_words, cand_words, lcs_len):
    """Task 3: Calculate word-level Precision, Recall, and F1."""
    r = len(ref_words)
    c = len(cand_words)

    if lcs_len == 0 or r == 0 or c == 0:
        return 0.0, 0.0, 0.0

    precision = lcs_len / c
    recall = lcs_len / r
    f1 = (2.0 * precision * recall) / (precision + recall)
    return precision, recall, f1


def run_batch_evaluation(pairs):
    """Task 4: Evaluate K summary pairs and identify the highest F1 pair."""
    best_pair_idx = -1
    best_f1 = -1.0

    for idx, (ref_text, cand_text) in enumerate(pairs, 1):
        print(f"\n==================== PAIR {idx} ====================")
        print(f"Reference : \"{ref_text}\"")
        print(f"Candidate : \"{cand_text}\"")

        ref_words = preprocess_words(ref_text)
        cand_words = preprocess_words(cand_text)

        L = compute_lcs_dp(ref_words, cand_words)
        lcs_words = reconstruct_lcs(ref_words, cand_words, L)
        lcs_len = L[len(ref_words)][len(cand_words)]

        # Verify against Self-Learning O(min(m, n)) two-row implementation
        opt_len = lcs_space_optimized(ref_words, cand_words)
        assert lcs_len == opt_len, "Mismatch between 2D DP and Two-Row LCS!"

        print_dp_table(ref_words, cand_words, L)
        prec, rec, f1 = compute_rouge_l(ref_words, cand_words, lcs_len)

        print(f"LCS length : {lcs_len} (2-Row Verified: {opt_len})")
        print(f"LCS words  : {' '.join(lcs_words) if lcs_words else 'NONE'}")
        print(f"Precision  : {prec:.3f}")
        print(f"Recall     : {rec:.3f}")
        print(f"F1         : {f1:.3f}")

        if f1 > best_f1:
            best_f1 = f1
            best_pair_idx = idx

    print("\n================================================")
    print(f"Best pair  : {best_pair_idx} (Highest F1 = {best_f1:.3f})")
    print("================================================")


def main():
    # Sample Test Case 1: Character-level verification
    print("=== SAMPLE TEST CASE 1: CHARACTER SEQUENCES ===")
    X_chars = list("ABCBDAB")
    Y_chars = list("BDCABA")
    L_char = compute_lcs_dp(X_chars, Y_chars)
    lcs_char = reconstruct_lcs(X_chars, Y_chars, L_char)
    print_dp_table(X_chars, Y_chars, L_char)
    print(f"LCS length : {L_char[len(X_chars)][len(Y_chars)]}")
    print(f"One LCS    : {''.join(lcs_char)}")

    # Sample Test Case 2 + 5 Self-Designed Test Cases (Task 4 & 5)
    print("\n=== BATCH SUMMARY EVALUATION (SAMPLE + 5 EDGE/TEST CASES) ===")
    summary_pairs = [
        # Pair 1: Sample Test Case 2
        ("the cat sat on the mat", "the cat is on the mat today"),
        # Pair 2: Identical summaries (F1 = 1.000)
        ("The court dismissed the appeal due to lack of evidence.",
         "The court dismissed the appeal due to lack of evidence!"),
        # Pair 3: Completely different summaries (F1 = 0.000)
        ("heavy rain flooded coastal districts",
         "stock markets rallied after interest rate cuts"),
        # Pair 4: Empty candidate summary edge case
        ("summary with valid legal text", ""),
        # Pair 5: Repeated words test case
        ("the the cat sat on the the mat", "the cat on the mat"),
        # Pair 6: Realistic news/legal summary evaluation
        ("Supreme court ruled that data privacy is a fundamental right under the constitution.",
         "The supreme court ruled data privacy is a fundamental right.")
    ]
    run_batch_evaluation(summary_pairs)


if __name__ == "__main__":
    main()
```

## Self Learning Part

### 1. Space-Optimized LCS ($\mathcal{O}(\min(m, n))$ Space)
In the recurrence $L[i][j] = \max(L[i-1][j], L[i][j-1])$ or $L[i-1][j-1] + 1$, computing row $i$ depends exclusively on row $i-1$. By storing only two 1D arrays (`prev` and `curr`) sized to $\min(m, n) + 1$, memory usage drops from quadratic $\mathcal{O}(m \times n)$ to linear $\mathcal{O}(\min(m, n))$ while producing the identical LCS length.

### 2. Comparison: LCS vs. Longest Common Substring vs. Edit Distance

| Metric / Problem | Ordering & Contiguity Rule | Allowed Operations | Small Example ($X = \text{"cat"}$, $Y = \text{"cart"}$) |
| :--- | :--- | :--- | :--- |
| **Longest Common Subsequence (LCS)** | Preserves relative order; need not be contiguous. | Skip/delete non-matching elements from either sequence. | Length = 3 ("cat", skipping 'r'). |
| **Longest Common Substring** | Must preserve order and be strictly contiguous. | Resets count to $0$ immediately upon any mismatch. | Length = 2 ("ca", broken by 'r'). |
| **Edit Distance (Levenshtein)** | Transforms $X$ into $Y$ using minimum operations. | Insertion, Deletion, and Substitution (each costing 1). | Distance = 1 (insert 'r' into "cat"). |

### 3. Reflection: Why Two-Row DP Cannot Backtrack & Hirschberg's Algorithm

* **Why Two Rows Fail Reconstruction:** Reconstructing the actual LCS requires tracing parent transitions backward from $(m, n)$ to $(0, 0)$ across all intermediate rows. The two-row optimization overwrites rows $0$ through $m-2$ as it advances, destroying the historical decision trail needed for backtracking.
* **Hirschberg's Algorithm:** Combines Divide and Conquer with two-row linear-space DP. It splits sequence $X$ at its midpoint $m/2$, runs forward two-row DP on $X[0..m/2]$ and backward two-row DP on $X[m/2..m]$, finds the optimal split index $j^*$ in $Y$ that maximizes the combined LCS length, and recurses on both halves—reconstructing the full LCS in $\mathcal{O}(m \times n)$ time and $\mathcal{O}(\min(m, n))$ space.

### 4. Additional Real-World Applications of LCS

* **Version Control & File Comparison (`git diff`):** Identifies unchanged lines between two versions of source code so additions and deletions can be highlighted minimally.
* **Bioinformatics & Genomics:** Aligns DNA, RNA, or protein sequences across species to detect conserved evolutionary regions and genetic mutations.
